```julia
using Plots
using Random
using Printf
using Base64
using Statistics

gr()

rastrigin(x::Vector{Float64}) =
    20.0 + x[1]^2 - 10cos(2π*x[1]) + x[2]^2 - 10cos(2π*x[2])

rosenbrock(x::Vector{Float64}) =
    100(x[2] - x[1]^2)^2 + (1 - x[1])^2

schwefel(x::Vector{Float64}) =
    418.9829*2 - x[1]*sin(sqrt(abs(x[1]))) - x[2]*sin(sqrt(abs(x[2])))

CONFIGS = [
    (rastrigin,  "Растригина",  (-5.12,   5.12),   [0.0,      0.0      ]),
    (rosenbrock, "Розенброка",  (-2.048,  2.048),  [1.0,      1.0      ]),
    (schwefel,   "Швефеля",     (-500.0,  500.0),  [420.9687, 420.9687 ]),
]

mutable struct Particle
    pos    :: Vector{Float64}
    vel    :: Vector{Float64}
    p_best :: Vector{Float64}
    pb_val :: Float64
end

function init_swarm_pso(f, dim, num_particles, lb, ub)
    particles = [
        let pos = lb .+ rand(dim) .* (ub - lb),
            vel = (rand(dim) .* 2 .- 1) .* (ub - lb) .* 0.1,
            val = f(pos)
            Particle(pos, vel, copy(pos), val)
        end
        for _ in 1:num_particles
    ]
    gi         = argmin([p.pb_val for p in particles])
    g_best     = copy(particles[gi].p_best)
    g_best_val = particles[gi].pb_val
    return particles, g_best, g_best_val
end

function pso_classic(f, dim;
                     num_particles :: Int     = 30,
                     max_iter      :: Int     = 60,
                     w             :: Float64 = 0.729,
                     c1            :: Float64 = 1.496,
                     c2            :: Float64 = 1.496,
                     bounds                  = (-5.12, 5.12))
    lb, ub = Float64(bounds[1]), Float64(bounds[2])
    particles, g_best, g_best_val = init_swarm_pso(f, dim, num_particles, lb, ub)
    all_positions = [hcat([copy(p.pos) for p in particles]...)]
    all_gbest     = [copy(g_best)]
    conv_history  = [g_best_val]
    for _ in 1:max_iter
        for p in particles
            rp = rand(dim); rg = rand(dim)
            p.vel = w  .* p.vel .+
                    c1 .* rp .* (p.p_best .- p.pos) .+
                    c2 .* rg .* (g_best   .- p.pos)
            p.pos .+= p.vel
            p.pos  .= clamp.(p.pos, lb, ub)
            val = f(p.pos)
            if val < p.pb_val; p.p_best = copy(p.pos); p.pb_val = val; end
        end
        bi = argmin([p.pb_val for p in particles])
        if particles[bi].pb_val < g_best_val
            g_best = copy(particles[bi].p_best); g_best_val = particles[bi].pb_val
        end
        push!(all_positions, hcat([copy(p.pos) for p in particles]...))
        push!(all_gbest, copy(g_best))
        push!(conv_history, g_best_val)
    end
    return all_positions, all_gbest, conv_history
end

mutable struct Fish
    pos :: Vector{Float64}
    val :: Float64
end

dist2(a, b) = sqrt(sum((a .- b).^2))

function init_fish(f, dim, N, lb, ub)
    [let pos = lb .+ rand(dim) .* (ub - lb)
         Fish(copy(pos), f(pos))
     end for _ in 1:N]
end

function random_move(xi_pos, dim, lb, ub, vision)
    y = copy(xi_pos)
    for k in 1:dim
        r = rand(); delta = rand()
        if r < 0.5
            y[k] = xi_pos[k] + delta * min(vision, ub - xi_pos[k])
        else
            y[k] = xi_pos[k] - delta * min(vision, xi_pos[k] - lb)
        end
    end
    return clamp.(y, lb, ub)
end

function leap!(fish::Fish, f, dim, lb, ub)
    for k in 1:dim
        r = rand(); delta = rand()
        if r < 0.5
            fish.pos[k] = fish.pos[k] + delta * (ub - fish.pos[k])
        else
            fish.pos[k] = fish.pos[k] - delta * (fish.pos[k] - lb)
        end
    end
    clamp!(fish.pos, lb, ub)
    fish.val = f(fish.pos)
end

function teleport!(fish::Fish, f, dim, lb, ub)
    fish.pos = lb .+ rand(dim) .* (ub - lb)
    fish.val = f(fish.pos)
end

function local_search(f, center, dim, lb, ub, sigma)
    best_pos = copy(center)
    best_val = f(center)
    for _ in 1:10
        y   = clamp.(center .+ sigma .* randn(dim), lb, ub)
        val = f(y)
        if val < best_val
            best_pos = y; best_val = val
        end
    end
    return best_pos, best_val
end

function afsa(f, dim;
              N              :: Int     = 30,
              max_iter       :: Int     = 60,
              vision_start   :: Float64 = 0.25,
              vision_min     :: Float64 = 0.05,
              crowd_thresh   :: Float64 = 0.618,
              stag_window    :: Int     = 5,
              stag_eps       :: Float64 = 1e-5,
              scout_prob     :: Float64 = 0.15,
              leaper_frac    :: Float64 = 0.20,
              bounds                   = (-5.12, 5.12))

    lb, ub = Float64(bounds[1]), Float64(bounds[2])
    span   = ub - lb

    fish = init_fish(f, dim, N, lb, ub)

    best_idx   = argmin([fi.val for fi in fish])
    g_best_val = fish[best_idx].val
    g_best     = copy(fish[best_idx].pos)

    all_positions = [hcat([copy(fi.pos) for fi in fish]...)]
    all_gbest     = [copy(g_best)]
    conv_history  = [g_best_val]
    best_history  = Float64[g_best_val]

    n_leapers    = max(1, round(Int, N * leaper_frac))
    stag_counter = 0

    for t in 1:max_iter
        frac   = vision_start - (vision_start - vision_min) * (t - 1) / (max_iter - 1)
        vision = frac * span

        for i in 1:N
            xi = fish[i]

            if rand() < scout_prob
                y   = random_move(xi.pos, dim, lb, ub, vision)
                val = f(y)
                if val < xi.val; xi.pos = y; xi.val = val; end
                continue
            end

            Vi = [j for j in 1:N if j != i && dist2(xi.pos, fish[j].pos) <= vision]

            if isempty(Vi)
                y   = random_move(xi.pos, dim, lb, ub, vision)
                val = f(y)
                if val < xi.val; xi.pos = y; xi.val = val; end
            else
                crowded = length(Vi) / N > crowd_thresh

                if crowded
                    j   = Vi[rand(1:length(Vi))]
                    r   = rand()
                    y   = clamp.(xi.pos .+ r .* (fish[j].pos .- xi.pos), lb, ub)
                    val = f(y)
                    if val < xi.val; xi.pos = y; xi.val = val; end
                else
                    ci     = mean([fish[j].pos for j in Vi])
                    fc     = f(ci)
                    j_star = Vi[argmin([fish[j].val for j in Vi])]
                    fj_s   = fish[j_star].val

                    if fc < xi.val
                        r   = rand()
                        y   = clamp.(xi.pos .+ r .* (ci .- xi.pos), lb, ub)
                        val = f(y)
                        if val < xi.val; xi.pos = y; xi.val = val; end
                    elseif fj_s < xi.val
                        r   = rand()
                        y   = clamp.(xi.pos .+ r .* (fish[j_star].pos .- xi.pos), lb, ub)
                        val = f(y)
                        if val < xi.val; xi.pos = y; xi.val = val; end
                    else
                        j   = Vi[rand(1:length(Vi))]
                        r   = rand()
                        y   = clamp.(xi.pos .+ r .* (fish[j].pos .- xi.pos), lb, ub)
                        val = f(y)
                        if val < xi.val; xi.pos = y; xi.val = val; end
                    end
                end
            end
        end

        bi = argmin([fi.val for fi in fish])
        if fish[bi].val < g_best_val
            g_best     = copy(fish[bi].pos)
            g_best_val = fish[bi].val
        end
        push!(best_history, g_best_val)

        stagnant = false
        if length(best_history) > stag_window
            ref      = best_history[end - stag_window]
            cur      = best_history[end]
            stagnant = (ref - cur) / max(abs(ref), 1e-12) < stag_eps
        end

        stag_counter = stagnant ? stag_counter + 1 : 0

        if stag_counter >= 1
            sigma       = vision * exp(-0.3 * stag_counter)
            new_pos, new_val = local_search(f, g_best, dim, lb, ub, sigma)
            if new_val < g_best_val
                g_best     = new_pos
                g_best_val = new_val
                worst_i    = argmax([fi.val for fi in fish])
                fish[worst_i].pos = copy(g_best)
                fish[worst_i].val = g_best_val
            end
        end

        if stag_counter >= 3
            idxs = randperm(N)[1:n_leapers]
            for j in idxs
                leap!(fish[j], f, dim, lb, ub)
                if fish[j].val < g_best_val
                    g_best     = copy(fish[j].pos)
                    g_best_val = fish[j].val
                end
            end
        end

        if stag_counter >= 8
            j = rand(1:N)
            teleport!(fish[j], f, dim, lb, ub)
            if fish[j].val < g_best_val
                g_best     = copy(fish[j].pos)
                g_best_val = fish[j].val
            end
        end

        push!(all_positions, hcat([copy(fi.pos) for fi in fish]...))
        push!(all_gbest, copy(g_best))
        push!(conv_history, g_best_val)
    end

    return all_positions, all_gbest, conv_history
end

function make_contour_grid(f, bounds; n=250)
    lb, ub = bounds
    xs = range(lb, ub, length=n)
    ys = range(lb, ub, length=n)
    Z  = [f([x, y]) for y in ys, x in xs]
    return xs, ys, clamp.(Z, quantile(vec(Z), 0.0), quantile(vec(Z), 0.97))
end

function legend_corner(true_min, bounds)
    lb, ub = bounds
    tm_nx = (true_min[1] - lb) / (ub - lb)
    tm_ny = (true_min[2] - lb) / (ub - lb)
    if tm_nx > 0.5 && tm_ny > 0.5;  return :bottomleft
    elseif tm_nx > 0.5;             return :topleft
    elseif tm_ny > 0.5;             return :bottomright
    else;                           return :topright
    end
end

function print_iter_table(label::String,
                           gbests::Vector{Vector{Float64}},
                           conv::Vector{Float64})
    n_iter = length(conv) - 1
    line   = repeat("-", 65)
    println()
    println("  ", label)
    println("  ", line)
    @printf("  %-6s  %14s  %14s  %18s\n", "iter", "x1", "x2", "f(g_best)")
    println("  ", line)
    prev_val = conv[1]
    for t in 0:n_iter
        gb       = gbests[t + 1]
        val      = conv[t + 1]
        improved = t > 0 && val < prev_val
        prev_val = val
        marker   = improved ? "  <-- новый минимум" : ""
        @printf("  %-6d  %14.6f  %14.6f  %18.8f%s\n",
                t, gb[1], gb[2], val, marker)
    end
    println("  ", line)
    @printf("  Итог: x* = (%.6f, %.6f)   f(x*) = %.8f\n",
            gbests[end][1], gbests[end][2], conv[end])
    println("  ", line)
    println()
end

function make_gif_b64(f, fname, mod_name, bounds, true_min,
                      positions, gbests, conv; fps=10)
    lb, ub    = bounds
    xs, ys, Z = make_contour_grid(f, bounds)
    n_iter    = length(positions) - 1
    conv_ymax = max(conv[1] * 1.05, 1e-6)
    lpos      = legend_corner(true_min, bounds)

    anim = @animate for t in 0:n_iter
        pos = positions[t + 1]; gb = gbests[t + 1]

        p_main = contourf(xs, ys, Z;
                          levels=25, color=:viridis, linewidth=0, alpha=0.88,
                          xlims=(lb,ub), ylims=(lb,ub), aspect_ratio=:equal,
                          xlabel="x1", ylabel="x2",
                          title=@sprintf("%s · %s   итерация %d/%d   f(g*) = %.5f",
                                         mod_name, fname, t, n_iter, conv[t+1]),
                          titlefontsize=9, guidefontsize=8, tickfontsize=7,
                          colorbar=false, legend=lpos, legendfontsize=7,
                          foreground_color_legend=:white,
                          background_color_legend=RGBA(0,0,0,0.55))

        scatter!(p_main, pos[1,:], pos[2,:];
                 label="рыба", color=:white,
                 markerstrokecolor=:black, markerstrokewidth=0.5, markersize=4)
        scatter!(p_main, [true_min[1]], [true_min[2]];
                 label="истинный min", color=:yellow, markershape=:diamond,
                 markersize=9, markerstrokecolor=:black, markerstrokewidth=1)
        scatter!(p_main, [gb[1]], [gb[2]];
                 label="g_best", color=:red, markershape=:star5,
                 markersize=10, markerstrokecolor=:white, markerstrokewidth=1)

        p_conv = plot(0:t, conv[1:t+1];
                      xlabel="итерация", ylabel="f(g_best)",
                      title=@sprintf("сходимость · %s · %s", mod_name, fname),
                      color=:steelblue, linewidth=2, legend=false,
                      xlims=(0,n_iter), ylims=(0,conv_ymax),
                      titlefontsize=8, guidefontsize=8, tickfontsize=7)

        plot(p_main, p_conv; layout=@layout([a{0.72w} b]), size=(900,440))
    end

    tmp = tempname() * ".gif"
    gif(anim, tmp; fps=fps, show_msg=false)
    b64 = open(io -> base64encode(read(io)), tmp, "r")
    rm(tmp)
    return b64
end

function make_png_b64(p)
    tmp = tempname() * ".png"
    savefig(p, tmp)
    b64 = open(io -> base64encode(read(io)), tmp, "r")
    rm(tmp)
    return b64
end

function display_gif(b64::String)
    display(HTML("""<img src="data:image/gif;base64,$(b64)" style="max-width:100%;display:block;margin:4px 0"/>"""))
end

function display_png(b64::String)
    display(HTML("""<img src="data:image/png;base64,$(b64)" style="max-width:100%;display:block;margin:4px 0"/>"""))
end

Random.seed!(42)

MAX_ITER   = 60
NUM_AGENTS = 35

for (f, fname, bounds, true_min) in CONFIGS
    positions, gbests, conv = afsa(f, 2;
        N            = NUM_AGENTS,
        max_iter     = MAX_ITER,
        vision_start = 0.25,
        vision_min   = 0.05,
        bounds       = bounds)

    print_iter_table("AFSA · Функция $(fname)", gbests, conv)

    b64 = make_gif_b64(f, fname, "AFSA", bounds, true_min,
                       positions, gbests, conv; fps=10)
    display_gif(b64)
end

println()
println(repeat("=", 65))
println("  СРАВНЕНИЕ AFSA vs PSO (классический) — сходимость")
println(repeat("=", 65))

for (cfg_idx, cfg_fname) in [(1,"Растригина"), (2,"Розенброка"), (3,"Швефеля")]
    f, _, bounds, _ = CONFIGS[cfg_idx]

    Random.seed!(42)
    _, _, conv_afsa = afsa(f, 2;
        N            = NUM_AGENTS,
        max_iter     = MAX_ITER,
        vision_start = 0.25,
        vision_min   = 0.05,
        bounds       = bounds)

    Random.seed!(42)
    _, _, conv_pso = pso_classic(f, 2;
        num_particles = NUM_AGENTS,
        max_iter      = MAX_ITER,
        bounds        = bounds)

    p = plot(;
        title         = "AFSA vs PSO · функция $(cfg_fname)",
        xlabel        = "Итерация",
        ylabel        = "f(g_best)",
        size          = (820, 380),
        legend        = :topright,
        titlefontsize = 11, guidefontsize=9, tickfontsize=8)

    plot!(p, 0:length(conv_afsa)-1, conv_afsa;
          label="AFSA (косяк рыб)", color=:steelblue, linewidth=2.5)
    plot!(p, 0:length(conv_pso)-1, conv_pso;
          label="PSO (классический)", color=:orangered, linewidth=2.5,
          linestyle=:dash)

    display_png(make_png_b64(p))
end

println()
println(repeat("=", 65))
println("  ВЛИЯНИЕ ПОЛЯ ЗРЕНИЯ НА AFSA · Функция Швефеля")
println(repeat("=", 65))

vision_vals   = [0.10, 0.20, 0.25, 0.40, 0.60]
vision_colors = [:purple, :steelblue, :mediumseagreen, :orange, :red]
vision_labels = ["10%", "20%", "25% (базовый)", "40%", "60%"]

f_sw, _, bounds_sw, _ = CONFIGS[3]

p_vis = plot(;
    title         = "Влияние поля зрения · функция Швефеля",
    xlabel        = "Итерация", ylabel="f(g_best)",
    size          = (820,380), legend=:topright,
    titlefontsize = 11, guidefontsize=9, tickfontsize=8)

for (i, vs) in enumerate(vision_vals)
    Random.seed!(42)
    _, _, conv = afsa(f_sw, 2;
        N            = NUM_AGENTS,
        max_iter     = MAX_ITER,
        vision_start = vs,
        vision_min   = 0.05,
        bounds       = bounds_sw)
    plot!(p_vis, 0:length(conv)-1, conv;
          label="поле зрения = $(vision_labels[i])", color=vision_colors[i], linewidth=2)
end
display_png(make_png_b64(p_vis))

println()
println(repeat("=", 65))
println("  ВЛИЯНИЕ ЧИСЛА РЫБ НА AFSA · Функция Швефеля")
println(repeat("=", 65))

swarm_sizes  = [10, 20, 35, 100]
swarm_colors = [:red, :orange, :steelblue, :mediumseagreen]

p_n = plot(;
    title         = "Влияние числа рыб · функция Швефеля",
    xlabel        = "Итерация", ylabel="f(g_best)",
    size          = (820,380), legend=:topright,
    titlefontsize = 11, guidefontsize=9, tickfontsize=8)

for (i, n) in enumerate(swarm_sizes)
    Random.seed!(42)
    _, _, conv = afsa(f_sw, 2;
        N            = n,
        max_iter     = MAX_ITER,
        vision_start = 0.25,
        vision_min   = 0.05,
        bounds       = bounds_sw)
    plot!(p_n, 0:length(conv)-1, conv;
          label="N = $n", color=swarm_colors[i], linewidth=2)
end
display_png(make_png_b64(p_n))

println()
println(repeat("=", 65))
println("  ВЛИЯНИЕ ВЕРОЯТНОСТИ РАЗВЕДКИ НА AFSA · Функция Швефеля")
println(repeat("=", 65))

scout_vals   = [0.0, 0.05, 0.15, 0.25, 0.40]
scout_colors = [:black, :blue, :mediumseagreen, :orange, :red]
scout_labels = ["0% — нет", "5%", "15% (базовый)", "25%", "40%"]

p_rp = plot(;
    title         = "Влияние вероятности разведки · функция Швефеля",
    xlabel        = "Итерация", ylabel="f(g_best)",
    size          = (820,380), legend=:topright,
    titlefontsize = 11, guidefontsize=9, tickfontsize=8)

for (i, sp) in enumerate(scout_vals)
    Random.seed!(42)
    _, _, conv = afsa(f_sw, 2;
        N            = NUM_AGENTS,
        max_iter     = MAX_ITER,
        vision_start = 0.25,
        vision_min   = 0.05,
        scout_prob   = sp,
        bounds       = bounds_sw)
    plot!(p_rp, 0:length(conv)-1, conv;
          label=scout_labels[i], color=scout_colors[i], linewidth=2)
end
display_png(make_png_b64(p_rp))
```
