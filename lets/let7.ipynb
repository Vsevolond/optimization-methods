```julia
using Plots
using Base64
using Printf
using Random

gr()

f(x) = 5 - 24x + 17x^2 - (11/3)*x^3 + (1/4)*x^4

TRUE_MIN_X = 1.0

function int_to_bits(x::Int, nbits::Int)
    bits = zeros(Int, nbits)
    tmp  = x
    for i in nbits:-1:1
        bits[i] = tmp % 2
        tmp     = tmp ÷ 2
    end
    return bits
end

function bits_to_int(bits::Vector{Int})
    val = 0
    n   = length(bits)
    for i in 1:n
        val += bits[i] * 2^(n - i)
    end
    return val
end

function crossover(p1::Vector{Int}, p2::Vector{Int})
    point  = rand(1:length(p1)-1)
    child1 = vcat(p1[1:point], p2[point+1:end])
    child2 = vcat(p2[1:point], p1[point+1:end])
    return child1, child2
end

function mutate(chrom::Vector{Int}, p_mut::Float64)
    rand() < p_mut || return copy(chrom)
    new_c = copy(chrom)
    new_c[rand(1:length(chrom))] ⊻= 1
    return new_c
end

function make_encoder(nbits::Int, lb::Float64, ub::Float64, step::Float64)
    max_idx = round(Int, (ub - lb) / step)
    encode  = (x::Float64) -> int_to_bits(round(Int, clamp(x, lb, ub) / step), nbits)
    decode  = (bits::Vector{Int}) -> round(min(bits_to_int(bits), max_idx) * step + lb, digits=6)
    return encode, decode
end

function ga_step(current_pop, n_pop, p_mut, fitness_fn)
    n           = length(current_pop)
    partner_ids = rand(1:n, n)
    offspring   = [crossover(current_pop[i], current_pop[partner_ids[i]])[1] for i in 1:n]
    mutated     = [mutate(ch, p_mut) for ch in offspring]
    all_inds    = vcat(current_pop, mutated)
    new_pop     = all_inds[sortperm(fitness_fn.(all_inds))[1:n_pop]]
    return new_pop, mutated
end

function print_header(title, params)
    sep = "=" ^ 60
    println(sep)
    println("  $title")
    println("  $params")
    println(sep)
end

function print_init_population(vals, fitness, d)
    println("Начальная популяция:")
    for (i, x) in enumerate(vals)
        println("  [$i]  x = $(round(x, digits=d))   f = $(round(fitness(x), digits=2))")
    end
end

function run_binary_ga(n_generations, p_mut)
    nbits     = 3
    n_pop     = 4
    population = [int_to_bits(v, nbits) for v in [2, 3, 5, 4]]
    decode    = bits_to_int
    fitness   = x -> f(Float64(bits_to_int(x)))

    print_header(
        "БИНАРНАЯ РЕАЛИЗАЦИЯ",
        "3 бита  |  Особей: $n_pop  |  Поколений: $n_generations  |  Диапазон [0, 7]"
    )
    println("Начальная популяция:  $(bits_to_int.(population))")
    println("Хромосомы:            $population")
    println("Приспособленность:    $(round.(f.(bits_to_int.(population)), digits=2))")

    history     = [(generation=0, initial=copy(population),
                    inter=Vector{Vector{Int}}(), final_pop=copy(population))]
    current_pop = copy(population)

    for gen in 1:n_generations
        initial_this_gen    = copy(current_pop)
        new_pop, mutated    = ga_step(current_pop, n_pop, p_mut, fitness)
        vals_new            = bits_to_int.(new_pop)

        println("\nПоколение $gen:")
        println("  Популяция:         $vals_new")
        println("  Приспособленность: $(round.(f.(vals_new), digits=2))")

        push!(history, (generation=gen, initial=initial_this_gen,
                        inter=mutated, final_pop=copy(new_pop)))
        current_pop = new_pop
    end

    best = bits_to_int(current_pop[1])
    println("\nЛучшее решение:  x = $best,  f(x) = $(round(f(best), digits=4))\n")
    return history, x -> Float64(bits_to_int(x))
end

function run_real_ga(n_generations, p_mut, n_pop, nbits, lb, ub, step, label; seed::Int=42)
    Random.seed!(seed)

    encode, decode = make_encoder(nbits, lb, ub, step)
    fitness        = x -> f(decode(x))
    d              = step >= 0.01 ? 2 : 6

    init_x     = [2.0, 3.0, 5.0, 4.0]
    extra      = [lb + rand() * (ub - lb) for _ in 1:(n_pop - 4)]
    population = [encode(x) for x in vcat(init_x, extra)]

    print_header(
        "ВЕЩЕСТВЕННАЯ РЕАЛИЗАЦИЯ — $label",
        "Битов: $nbits  |  Шаг: $step  |  Особей: $n_pop  |  Поколений: $n_generations  |  Вер. мутации: $p_mut"
    )
    print_init_population(decode.(population), x -> f(x), d)

    history     = [(generation=0, initial=copy(population),
                    inter=Vector{Vector{Int}}(), final_pop=copy(population))]
    current_pop = copy(population)

    for gen in 1:n_generations
        initial_this_gen = copy(current_pop)
        new_pop, mutated = ga_step(current_pop, n_pop, p_mut, fitness)
        best_x           = decode(new_pop[1])

        println("\nПоколение $gen:")
        println("  Лучший x = $(round(best_x, digits=d))  →  f = $(round(f(best_x), digits=4))")

        push!(history, (generation=gen, initial=initial_this_gen,
                        inter=mutated, final_pop=copy(new_pop)))
        current_pop = new_pop
    end

    best = decode(current_pop[1])
    println("\nЛучшее решение:  x = $(round(best, digits=d)),  f(x) = $(round(f(best), digits=d+2))\n")
    return history, decode
end

function make_gif(history, title_prefix, decode_fn, fps::Int)
    x_range = range(-1, 9, length=400)
    y_range = f.(x_range)

    anim = @animate for state in history
        gen        = state.generation
        init_vals  = Float64[clamp(decode_fn(x), 0.0, 7.0) for x in state.initial]
        inter_vals = Float64[clamp(decode_fn(x), 0.0, 7.0) for x in state.inter]
        fin_vals   = Float64[clamp(decode_fn(x), 0.0, 7.0) for x in state.final_pop]
        best_val   = minimum(f.(fin_vals))

        plot(x_range, y_range,
            label   = "f(x)", color = :darkgreen, lw = 2,
            title   = "$title_prefix — Поколение $gen  |  f* = $(round(best_val, digits=4))",
            xlabel  = "x", ylabel = "f(x)",
            xlims   = (-1, 9), ylims = (-30, 100),
            legend  = :topright, size = (750, 480))

        vline!([TRUE_MIN_X],
            color = :purple, lw = 1.5, linestyle = :dash,
            label = "x* ≈ $(TRUE_MIN_X)")

        scatter!(init_vals, f.(init_vals),
            marker = :cross, markersize = 10, color = :blue,
            markerstrokewidth = 2, markerstrokecolor = :blue,
            label  = "Начальные особи")

        !isempty(inter_vals) && scatter!(inter_vals, f.(inter_vals),
            marker = :circle, markersize = 6, color = :red,
            label  = "Промежуточные особи")

        scatter!(fin_vals, f.(fin_vals),
            marker = :circle, markersize = 9, color = :black,
            label  = "Финальные особи")
    end

    tmp = tempname() * ".gif"
    gif(anim, tmp; fps=fps, show_msg=false)
    b64 = open(io -> base64encode(read(io)), tmp, "r")
    rm(tmp)
    return b64
end

function display_gif(b64::String)
    display(HTML("""<img src="data:image/gif;base64,$(b64)"
        style="max-width:100%;display:block;margin:8px 0"/>"""))
end

p_mut = 0.3

Random.seed!(42)
history_bin, decode_bin_fn = run_binary_ga(10, p_mut)
display_gif(make_gif(history_bin, "ГА (бинарный)", decode_bin_fn, 1))

history_r1, decode_r1 = run_real_ga(10, p_mut, 4, 10, 0.0, 7.0, 0.01,
                                     "4 особи / 10 бит / шаг 0.01 / 10 поколений"; seed=42)
display_gif(make_gif(history_r1, "ГА (вещественный)", decode_r1, 1))

history_r2, decode_r2 = run_real_ga(30, p_mut, 20, 10, 0.0, 7.0, 0.01,
                                     "20 особей / 10 бит / шаг 0.01 / 30 поколений"; seed=42)
display_gif(make_gif(history_r2, "ГА (вещественный, больше особей)", decode_r2, 2))

history_r3, decode_r3 = run_real_ga(50, p_mut, 20, 20, 0.0, 7.0,
                                     7.0 / (2^20 - 1),
                                     "20 особей / 20 бит / 50 поколений"; seed=42)
display_gif(make_gif(history_r3, "ГА (вещественный, высокая точность)", decode_r3, 5))
```
