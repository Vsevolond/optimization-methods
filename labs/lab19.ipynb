```julia
using Random, Statistics, LinearAlgebra, Printf, Plots, Base64

gr()

const N_COMPLETE = 5

function build_distance_matrix(pts; rounding=true)
    n = length(pts)
    D = zeros(Float64, n, n)
    for i in 1:n, j in 1:n
        i == j && continue
        d = hypot(pts[i][1] - pts[j][1], pts[i][2] - pts[j][2])
        D[i, j] = rounding ? round(d) : d
    end
    D
end

function tour_length(tour::Vector{Int}, D::Matrix{Float64})
    s = 0.0
    n = length(tour)
    @inbounds for k in 1:n-1
        s += D[tour[k], tour[k+1]]
    end
    s + D[tour[end], tour[1]]
end

function nearest_neighbor_tour(D::Matrix{Float64}; start::Int=1)
    n = size(D, 1)
    visited = falses(n)
    tour = Int[start]
    visited[start] = true
    cur = start
    for _ in 2:n
        nxt = 0
        best = Inf
        for j in 1:n
            visited[j] && continue
            if D[cur, j] < best
                best = D[cur, j]
                nxt = j
            end
        end
        push!(tour, nxt)
        visited[nxt] = true
        cur = nxt
    end
    return tour, tour_length(tour, D)
end

function best_nearest_neighbor_tour(D::Matrix{Float64})
    n = size(D, 1)
    best_tour = Int[]
    best_len = Inf
    for s in 1:n
        t, L = nearest_neighbor_tour(D; start=s)
        if L < best_len
            best_tour = t
            best_len = L
        end
    end
    return best_tour, best_len
end

function delta_2opt(tour::Vector{Int}, D::Matrix{Float64}, i::Int, j::Int)
    n = length(tour)
    (i == 1 && j == n) && return 0.0
    a = tour[i]
    b = tour[mod1(i + 1, n)]
    c = tour[j]
    d = tour[mod1(j + 1, n)]
    (D[a, c] + D[b, d]) - (D[a, b] + D[c, d])
end

function apply_2opt!(tour::Vector{Int}, i::Int, j::Int)
    tour[i+1:j] = reverse(tour[i+1:j])
    return tour
end

function two_opt_best_improvement!(tour::Vector{Int}, D::Matrix{Float64}; max_passes=200)
    n = length(tour)
    L = tour_length(tour, D)

    for _ in 1:max_passes
        bestΔ = 0.0
        best_i, best_j = 0, 0

        for i in 1:n-1, j in i+2:n
            (i == 1 && j == n) && continue
            Δ = delta_2opt(tour, D, i, j)
            if Δ < bestΔ
                bestΔ = Δ
                best_i, best_j = i, j
            end
        end

        if bestΔ < -1e-12
            apply_2opt!(tour, best_i, best_j)
            L += bestΔ
        else
            break
        end
    end

    return tour, L
end

function two_opt_strong!(tour::Vector{Int}, D::Matrix{Float64})
    return two_opt_best_improvement!(tour, D; max_passes=300)
end

function insertion_move(tour::Vector{Int}, i::Int, j::Int)
    n = length(tour)
    i == j && return copy(tour)
    t = copy(tour)
    city = t[i]
    deleteat!(t, i)
    if j > i
        insert!(t, j - 1, city)
    else
        insert!(t, j, city)
    end
    length(t) == n || error("Insertion move broke permutation")
    return t
end

function double_bridge_move(tour::Vector{Int}, rng)
    n = length(tour)
    cuts = sort(rand(rng, 2:n-1, 4))
    a, b, c, d = cuts
    p1 = tour[1:a]
    p2 = tour[a+1:b]
    p3 = tour[b+1:c]
    p4 = tour[c+1:d]
    p5 = tour[d+1:end]
    return vcat(p1, p3, p2, p4, p5)
end

function route_xy(pts, tour)
    cyc = vcat(tour, tour[1])
    xs = [pts[i][1] for i in cyc]
    ys = [pts[i][2] for i in cyc]
    return xs, ys
end

function gif_to_base64(tmpfile)
    b64 = open(io -> base64encode(read(io)), tmpfile, "r")
    rm(tmpfile; force=true)
    return b64
end

function display_side_by_side_gifs(left_b64, right_b64)
    html = """
    <div style="display:flex;gap:16px;align-items:flex-start;flex-wrap:nowrap;margin:8px 0 18px 0;">
      <img src="data:image/gif;base64,$left_b64" style="max-width:49%;height:auto;display:block;border:1px solid #ddd"/>
      <img src="data:image/gif;base64,$right_b64" style="max-width:49%;height:auto;display:block;border:1px solid #ddd"/>
    </div>
    """
    display(HTML(html))
end

make_geometric(α) = (T0, t, T) -> T * α
make_log(c)       = (T0, t, T) -> T0 / (1 + c * log(t + 2.0))
make_veryfast(β)  = (T0, t, T) -> T0 * exp(-β * t)

COLORS  = [:royalblue, :firebrick, :darkorange]
LSTYLES = [:solid, :dash, :dot]

function sa_tsp_single(
    D::Matrix{Float64}, T0, cool_fn, optimal_len;
    max_iter=450_000,
    k_stop=120_000,
    Tmin_frac=1e-7,
    seed=42,
    intensify_every=3000,
    strong_every=12000,
    reheat_every=18000,
    reheat_factor=0.65,
    save_stride=1500,
    use_doublebridge=true,
    start_mode=:best_nn,
    initial_strong_local_search=true
)
    rng = MersenneTwister(seed)
    n = size(D, 1)

    if start_mode == :best_nn
        x, L = best_nearest_neighbor_tour(D)
    elseif start_mode == :random
        x = randperm(rng, n)
        L = tour_length(x, D)
    else
        error("Unknown start_mode: $start_mode")
    end

    if initial_strong_local_search
        x, L = two_opt_strong!(x, D)
    end

    best_x = copy(x)
    best_val = L
    T = T0
    Tmin_stop = Tmin_frac * T0

    curr_hist = Vector{Float64}(undef, max_iter)
    best_hist = Vector{Float64}(undef, max_iter)
    T_hist    = Vector{Float64}(undef, max_iter)

    saved_tours      = Vector{Vector{Int}}()
    saved_best_tours = Vector{Vector{Int}}()
    saved_iter       = Int[]
    saved_curr       = Float64[]
    saved_best       = Float64[]

    no_improve = 0

    for t in 1:max_iter
        move_type = rand(rng)

        x_new = x
        L_new = L

        if move_type < 0.50
            i = rand(rng, 1:n-1)
            j = rand(rng, i+1:n)
            if !(i == 1 && j == n)
                Δ = delta_2opt(x, D, i, j)
                x_new = copy(x)
                apply_2opt!(x_new, i, j)
                L_new = L + Δ
            end

        elseif move_type < 0.75
            i = rand(rng, 1:n)
            j = rand(rng, 1:n)
            while j == i
                j = rand(rng, 1:n)
            end
            x_new = insertion_move(x, i, j)
            L_new = tour_length(x_new, D)

        elseif move_type < 0.90
            i = rand(rng, 1:n)
            j = rand(rng, 1:n)
            while j == i
                j = rand(rng, 1:n)
            end
            x_new = copy(x)
            x_new[i], x_new[j] = x_new[j], x_new[i]
            L_new = tour_length(x_new, D)

        else
            if use_doublebridge
                x_new = double_bridge_move(x, rng)
                L_new = tour_length(x_new, D)
            end
        end

        Δ = L_new - L

        if Δ <= 0 || rand(rng) < exp(-Δ / max(T, 1e-300))
            x = x_new
            L = L_new
        end

        if t % intensify_every == 0
            x2 = copy(x)
            x2, L2 = two_opt_best_improvement!(x2, D; max_passes=40)
            if L2 < L
                x = x2
                L = L2
            end
        end

        if t % strong_every == 0
            x2 = copy(x)
            x2, L2 = two_opt_strong!(x2, D)
            if L2 < L
                x = x2
                L = L2
            end
        end

        if L < best_val - 1e-12
            best_x = copy(x)
            best_val = L
            no_improve = 0
        else
            no_improve += 1
        end

        curr_hist[t] = L
        best_hist[t] = best_val
        T_hist[t]    = T

        if t == 1 || t % save_stride == 0 || t == max_iter
            push!(saved_tours, copy(x))
            push!(saved_best_tours, copy(best_x))
            push!(saved_iter, t)
            push!(saved_curr, L)
            push!(saved_best, best_val)
        end

        if t % reheat_every == 0 && no_improve > reheat_every ÷ 2 && best_val > optimal_len
            T = max(T, reheat_factor * T0)

            if use_doublebridge
                x = double_bridge_move(best_x, rng)
                x, L = two_opt_best_improvement!(x, D; max_passes=60)
            end

            if best_val <= optimal_len + max(150.0, 0.02 * optimal_len)
                x = double_bridge_move(best_x, rng)
                x, L = two_opt_strong!(x, D)
            end

            no_improve = 0
        else
            T = max(cool_fn(T0, t, T), 1e-12)
        end

        if best_val <= optimal_len + 1e-12
            if saved_iter[end] != t
                push!(saved_tours, copy(x))
                push!(saved_best_tours, copy(best_x))
                push!(saved_iter, t)
                push!(saved_curr, L)
                push!(saved_best, best_val)
            end
            return (
                best_tour        = best_x,
                best_len         = best_val,
                curr_hist        = curr_hist[1:t],
                best_hist        = best_hist[1:t],
                T_hist           = T_hist[1:t],
                saved_tours      = saved_tours,
                saved_best_tours = saved_best_tours,
                saved_iter       = saved_iter,
                saved_curr       = saved_curr,
                saved_best       = saved_best,
                iters_done       = t
            )
        end

        if no_improve >= k_stop && T <= Tmin_stop
            if saved_iter[end] != t
                push!(saved_tours, copy(x))
                push!(saved_best_tours, copy(best_x))
                push!(saved_iter, t)
                push!(saved_curr, L)
                push!(saved_best, best_val)
            end
            return (
                best_tour        = best_x,
                best_len         = best_val,
                curr_hist        = curr_hist[1:t],
                best_hist        = best_hist[1:t],
                T_hist           = T_hist[1:t],
                saved_tours      = saved_tours,
                saved_best_tours = saved_best_tours,
                saved_iter       = saved_iter,
                saved_curr       = saved_curr,
                saved_best       = saved_best,
                iters_done       = t
            )
        end
    end

    return (
        best_tour        = best_x,
        best_len         = best_val,
        curr_hist        = curr_hist,
        best_hist        = best_hist,
        T_hist           = T_hist,
        saved_tours      = saved_tours,
        saved_best_tours = saved_best_tours,
        saved_iter       = saved_iter,
        saved_curr       = saved_curr,
        saved_best       = saved_best,
        iters_done       = max_iter
    )
end

function construct_tour(start::Int, tau, eta, alpha, beta, n, rng)
    tour = [start]
    visited = falses(n)
    visited[start] = true
    current = start
    for _ in 1:n-1
        weights = zeros(Float64, n)
        total = 0.0
        for j in 1:n
            visited[j] && continue
            w = (tau[current, j]^alpha) * (eta[current, j]^beta)
            weights[j] = w
            total += w
        end
        nxt = 0
        if total <= 0
            candidates = [j for j in 1:n if !visited[j]]
            nxt = rand(rng, candidates)
        else
            r = rand(rng) * total
            acc = 0.0
            for j in 1:n
                visited[j] && continue
                acc += weights[j]
                if r <= acc
                    nxt = j
                    break
                end
            end
            nxt == 0 && (nxt = findfirst(!, visited))
        end
        push!(tour, nxt)
        visited[nxt] = true
        current = nxt
    end
    return tour
end

function run_aco(D::Matrix{Float64}, optimal_len;
                 num_ants=size(D, 1),
                 num_iters=300,
                 alpha=1.0,
                 beta=5.0,
                 rho=0.5,
                 Q=100.0,
                 tau0=1e-6,
                 elite_weight=3.0,
                 seed=42)

    rng = MersenneTwister(seed)
    n = size(D, 1)

    eta = zeros(Float64, n, n)
    for i in 1:n, j in 1:n
        i != j && (eta[i, j] = 1.0 / D[i, j])
    end

    tau = fill(tau0, n, n)
    for i in 1:n
        tau[i, i] = 0.0
    end

    best_tour = Int[]
    best_len  = Inf

    for _ in 1:num_iters
        iter_tours = Vector{Vector{Int}}()
        iter_lens  = Float64[]

        for _ in 1:num_ants
            tour = construct_tour(rand(rng, 1:n), tau, eta, alpha, beta, n, rng)
            tour, L = two_opt_best_improvement!(tour, D; max_passes=40)

            push!(iter_tours, tour)
            push!(iter_lens, L)

            if L < best_len
                best_len = L
                best_tour = copy(tour)
            end
        end

        tau .*= (1.0 - rho)

        for (tour, L) in zip(iter_tours, iter_lens)
            Δτ = Q / L
            for k in 1:n-1
                i, j = tour[k], tour[k+1]
                tau[i, j] += Δτ
                tau[j, i] += Δτ
            end
            i, j = tour[end], tour[1]
            tau[i, j] += Δτ
            tau[j, i] += Δτ
        end

        if !isempty(best_tour)
            Δe = elite_weight * Q / best_len
            for k in 1:n-1
                i, j = best_tour[k], best_tour[k+1]
                tau[i, j] += Δe
                tau[j, i] += Δe
            end
            i, j = best_tour[end], best_tour[1]
            tau[i, j] += Δe
            tau[j, i] += Δe
        end

        if best_len <= optimal_len + 1e-12
            return (
                best_tour   = best_tour,
                best_len    = best_len
            )
        end
    end

    return (
        best_tour   = best_tour,
        best_len    = best_len
    )
end

function draw_complete_graph(pts; title_str="Полносвязный граф")
    xs = [p[1] for p in pts]
    ys = [p[2] for p in pts]
    p = plot(; legend=false, aspect_ratio=:equal, xlabel="x", ylabel="y", title=title_str)
    n = length(pts)
    for i in 1:n-1, j in i+1:n
        plot!(p, [pts[i][1], pts[j][1]], [pts[i][2], pts[j][2]];
              color=:gray, alpha=0.45, linewidth=1.0, linestyle=:dot)
    end
    scatter!(p, xs, ys; color=:black, markersize=6)
    annotate!(p, [(pts[i][1], pts[i][2], text(string(i), 9, :darkblue)) for i in 1:n])
    return p
end

function exact_tsp_bruteforce(D::Matrix{Float64})
    n = size(D, 1)
    cities = collect(2:n)
    best_len = Inf
    best_tour = Int[]
    used = falses(length(cities))
    perm = Int[]

    function dfs!()
        if length(perm) == length(cities)
            tour = vcat(1, perm)
            L = tour_length(tour, D)
            if L < best_len
                best_len = L
                best_tour = copy(tour)
            end
            return
        end
        for i in eachindex(cities)
            used[i] && continue
            used[i] = true
            push!(perm, cities[i])
            dfs!()
            pop!(perm)
            used[i] = false
        end
    end

    dfs!()
    return best_tour, best_len
end

function draw_tour(pts, tour; title_str="", color=:royalblue, lw=2.0)
    xs, ys = route_xy(pts, tour)
    city_x = [p[1] for p in pts]
    city_y = [p[2] for p in pts]
    p = scatter(city_x, city_y;
                markersize=4, color=:black, label="города",
                xlabel="x", ylabel="y", aspect_ratio=:equal,
                title=title_str, legend=:topright)
    plot!(p, xs, ys; color=color, linewidth=lw, label="маршрут")
    return p
end

function make_tsp_route_gif(pts, saved_tours, saved_iter, saved_curr, saved_best, fname, sname, color;
                            fps=9, frame_idx=nothing, hold_last=8)
    city_x = [p[1] for p in pts]
    city_y = [p[2] for p in pts]
    xmin, xmax = minimum(city_x), maximum(city_x)
    ymin, ymax = minimum(city_y), maximum(city_y)
    padx = 0.06 * max(1.0, xmax - xmin)
    pady = 0.06 * max(1.0, ymax - ymin)

    N = length(saved_tours)
    frames = isnothing(frame_idx) ? collect(1:N) : collect(frame_idx)
    frames = clamp.(frames, 1, N)

    anim = @animate for k in vcat(frames, fill(frames[end], hold_last))
        tour = saved_tours[k]
        xs, ys = route_xy(pts, tour)

        p = scatter(city_x, city_y;
                    markersize=4, color=:black,
                    xlabel="x", ylabel="y",
                    title="$fname  |  $sname\nИтерация $(saved_iter[k])  |  current = $(round(saved_curr[k], digits=2))  |  best = $(round(saved_best[k], digits=2))",
                    legend=:topright, aspect_ratio=:equal,
                    xlims=(xmin-padx, xmax+padx), ylims=(ymin-pady, ymax+pady),
                    titlefontsize=9, guidefontsize=9, legendfontsize=7,
                    size=(620, 420), label="города")

        plot!(xs, ys; color=color, linewidth=2.2, alpha=0.95, label="текущий маршрут")
        scatter!([pts[tour[1]][1]], [pts[tour[1]][2]];
                 color=:green, markersize=7, markershape=:star5, label="старт")
    end

    tmp = tempname() * ".gif"
    gif(anim, tmp; fps=fps, show_msg=false)
    return gif_to_base64(tmp)
end

function make_best_route_gif(pts, saved_best_tours, saved_iter, saved_best, fname, sname, color;
                             fps=9, frame_idx=nothing, hold_last=8)
    city_x = [p[1] for p in pts]
    city_y = [p[2] for p in pts]
    xmin, xmax = minimum(city_x), maximum(city_x)
    ymin, ymax = minimum(city_y), maximum(city_y)
    padx = 0.06 * max(1.0, xmax - xmin)
    pady = 0.06 * max(1.0, ymax - ymin)

    N = length(saved_best_tours)
    frames = isnothing(frame_idx) ? collect(1:N) : collect(frame_idx)
    frames = clamp.(frames, 1, N)

    anim = @animate for k in vcat(frames, fill(frames[end], hold_last))
        tour = saved_best_tours[k]
        xs, ys = route_xy(pts, tour)

        p = scatter(city_x, city_y;
                    markersize=4, color=:black,
                    xlabel="x", ylabel="y",
                    title="$fname  |  $sname\nИтерация $(saved_iter[k])  |  best = $(round(saved_best[k], digits=2))",
                    legend=:topright, aspect_ratio=:equal,
                    xlims=(xmin-padx, xmax+padx), ylims=(ymin-pady, ymax+pady),
                    titlefontsize=9, guidefontsize=9, legendfontsize=7,
                    size=(620, 420), label="города")

        plot!(xs, ys; color=color, linewidth=2.8, alpha=0.98, label="лучший маршрут")
        scatter!([pts[tour[1]][1]], [pts[tour[1]][2]];
                 color=:green, markersize=7, markershape=:star5, label="старт")
    end

    tmp = tempname() * ".gif"
    gif(anim, tmp; fps=fps, show_msg=false)
    return gif_to_base64(tmp)
end

function draw_tour(pts, tour; title_str="", color=:royalblue, lw=2.0)
    xs, ys = route_xy(pts, tour)
    city_x = [p[1] for p in pts]
    city_y = [p[2] for p in pts]
    p = scatter(city_x, city_y;
                markersize=4, color=:black, label="города",
                xlabel="x", ylabel="y", aspect_ratio=:equal,
                title=title_str, legend=:topright)
    plot!(p, xs, ys; color=color, linewidth=lw, label="маршрут")
    return p
end

rng_small = MersenneTwister(16)
PTS_COMPLETE = [(round(100 * rand(rng_small), digits=2), round(100 * rand(rng_small), digits=2)) for _ in 1:N_COMPLETE]

display(draw_complete_graph(PTS_COMPLETE; title_str="Полносвязный граф из $N_COMPLETE точек"))

D_COMPLETE = build_distance_matrix(PTS_COMPLETE; rounding=true)
OPT_TOUR_COMPLETE, OPT_LEN_COMPLETE = exact_tsp_bruteforce(D_COMPLETE)

println()
println("Точный оптимальный маршрут: $OPT_TOUR_COMPLETE")
println("Точная оптимальная длина: $(round(OPT_LEN_COMPLETE, digits=2))")
println()

SCHEDULES_COMPLETE = [
    ("Геометрическое (α=0.985)",  220.0, make_geometric(0.985)),
    ("Логарифмическое (c=0.30)",  260.0, make_log(0.30)),
    ("Сверхбыстрый (β=0.008)",    180.0, make_veryfast(0.008)),
]

SA_PARAMS_COMPLETE = (8_000, 2_000, 1e-10, 120, 500, 700, 0.60, 1)

max_iter_c, k_stop_c, Tmin_frac_c, intensify_every_c, strong_every_c, reheat_every_c, reheat_factor_c, save_stride_c = SA_PARAMS_COMPLETE

println("▶  Полносвязный граф ($N_COMPLETE точек) — имитация отжига")
println("   max_iter=$max_iter_c   k_stop=$k_stop_c   Tmin_frac=$Tmin_frac_c")
println("   intensify_every=$intensify_every_c   strong_every=$strong_every_c   reheat_every=$reheat_every_c")
println("   " * "─"^72)

sa_results_complete = NamedTuple[]

pc_c = plot(title="Полносвязный граф ($N_COMPLETE точек) — текущее L на каждой итерации",
            xlabel="Итерация", ylabel="Текущее L",
            legend=:topright, titlefontsize=10, guidefontsize=8, legendfontsize=7)
hline!(pc_c, [OPT_LEN_COMPLETE]; label="L* = $(round(OPT_LEN_COMPLETE, digits=2))",
       color=:green, linestyle=:dashdot, linewidth=1.5)

pb_c = plot(title="Полносвязный граф ($N_COMPLETE точек) — лучшее L на каждой итерации",
            xlabel="Итерация", ylabel="Лучшее L",
            legend=:topright, titlefontsize=10, guidefontsize=8, legendfontsize=7)
hline!(pb_c, [OPT_LEN_COMPLETE]; label="L* = $(round(OPT_LEN_COMPLETE, digits=2))",
       color=:green, linestyle=:dashdot, linewidth=1.5)

for (k, (sname, T0, cool_fn)) in enumerate(SCHEDULES_COMPLETE)
    t_start = time()

    res = sa_tsp_single(
        D_COMPLETE, T0, cool_fn, OPT_LEN_COMPLETE;
        max_iter=max_iter_c,
        k_stop=k_stop_c,
        Tmin_frac=Tmin_frac_c,
        seed=142 + 100k,
        intensify_every=intensify_every_c,
        strong_every=strong_every_c,
        reheat_every=reheat_every_c,
        reheat_factor=reheat_factor_c,
        save_stride=save_stride_c,
        use_doublebridge=true,
        start_mode=:random,
        initial_strong_local_search=false
    )

    elapsed = time() - t_start
    gap = 100.0 * (res.best_len - OPT_LEN_COMPLETE) / OPT_LEN_COMPLETE

    @printf("   %-38s │ L*=%.2f  |Δ|=%.2f  │ %8d iter  │ %.4f с\n",
            sname, res.best_len, abs(res.best_len - OPT_LEN_COMPLETE),
            res.iters_done, elapsed)
    @printf("   %38s   gap = %+8.4f%%\n\n", "", gap)

    idx_curr = 1:length(res.curr_hist)
    idx_best = 1:length(res.best_hist)

    plot!(pc_c, idx_curr, res.curr_hist[idx_curr];
          label=sname * @sprintf(" → %.1f", res.best_len),
          color=COLORS[k], linestyle=LSTYLES[k], linewidth=1.4, alpha=0.85,
          marker=:circle, markersize=2)

    plot!(pb_c, idx_best, res.best_hist[idx_best];
          label=sname * @sprintf(" → %.1f", res.best_len),
          color=COLORS[k], linestyle=LSTYLES[k], linewidth=1.8, alpha=0.90,
          marker=:circle, markersize=1.8)

    push!(sa_results_complete, (
        sname   = sname,
        color   = COLORS[k],
        result  = res,
        elapsed = elapsed,
        gap     = gap
    ))
end

display(pc_c)
display(pb_c)

println()
for item in sa_results_complete
    println("   ── $(item.sname)")

    shared_idx = collect(1:length(item.result.saved_iter))

    curr_b64 = make_tsp_route_gif(
        PTS_COMPLETE,
        item.result.saved_tours,
        item.result.saved_iter,
        item.result.saved_curr,
        item.result.saved_best,
        "Граф из $N_COMPLETE точек", item.sname, item.color;
        fps=9,
        frame_idx=shared_idx,
        hold_last=10
    )

    best_b64 = make_best_route_gif(
        PTS_COMPLETE,
        item.result.saved_best_tours,
        item.result.saved_iter,
        item.result.saved_best,
        "Граф из $N_COMPLETE точек", item.sname, item.color;
        fps=9,
        frame_idx=shared_idx,
        hold_last=10
    )

    display_side_by_side_gifs(curr_b64, best_b64)
end

println()
println("▶  Полносвязный граф ($N_COMPLETE точек) — муравьиный алгоритм")
println("   " * "─"^72)

t_aco_c = time()
aco_c = run_aco(D_COMPLETE, OPT_LEN_COMPLETE;
                num_ants=max(20, N_COMPLETE * 4),
                num_iters=120,
                alpha=1.0,
                beta=4.0,
                rho=0.5,
                Q=50.0,
                tau0=1e-4,
                elite_weight=2.0,
                seed=24)
elapsed_aco_c = time() - t_aco_c
gap_aco_c = 100.0 * (aco_c.best_len - OPT_LEN_COMPLETE) / OPT_LEN_COMPLETE

@printf("   %-38s │ L*=%.2f  |Δ|=%.2f  │ %.4f с\n",
        "ACO + 2-opt", aco_c.best_len, abs(aco_c.best_len - OPT_LEN_COMPLETE), elapsed_aco_c)
@printf("   %38s   gap = %+8.4f%%\n",
        "", gap_aco_c)

display(draw_tour(PTS_COMPLETE, aco_c.best_tour;
                  title_str=@sprintf("Граф из %d точек — ACO+2opt   L=%.2f", N_COMPLETE, aco_c.best_len),
                  color=:purple, lw=2.0))

println()
@printf("  %-34s │ %-10s │ %-10s │ %-12s │ %-10s\n",
        "Метод", "Лучшее L", "gap,%", "iter", "time, с")
println("  " * "─"^34 * "┼" * "─"^12 * "┼" * "─"^12 * "┼" * "─"^14 * "┼" * "─"^12)

for item in sa_results_complete
    @printf("  %-34s │ %-10.2f │ %+9.4f │ %-12d │ %-10.4f\n",
            item.sname, item.result.best_len, item.gap,
            item.result.iters_done, item.elapsed)
end

@printf("  %-34s │ %-10.2f │ %+9.4f │ %-12s │ %-10.4f\n",
        "ACO + 2-opt", aco_c.best_len, gap_aco_c, "—", elapsed_aco_c)
println("=" ^ 86)

println()

best_len_sa_c = minimum(x.result.best_len for x in sa_results_complete)
cands_c = [x for x in sa_results_complete if abs(x.result.best_len - best_len_sa_c) < 1e-12]
best_sa_c = cands_c[argmin([x.elapsed for x in cands_c])]

println("Лучший вариант SA для графа из $N_COMPLETE точек : $(best_sa_c.sname)")
@printf("L = %.2f, gap = %+0.4f%%, итераций = %d, время = %.4f с\n",
        best_sa_c.result.best_len, best_sa_c.gap, best_sa_c.result.iters_done, best_sa_c.elapsed)
@printf("ACO + 2-opt: L = %.2f, gap = %+0.4f%%, время = %.4f с\n",
        aco_c.best_len, gap_aco_c, elapsed_aco_c)
```

```julia
using Random, Statistics, LinearAlgebra, Printf, Plots, Base64

gr()

const POINTS_BERLIN52 = [
    (565.0,575.0),(25.0,185.0),(345.0,750.0),(945.0,685.0),
    (845.0,655.0),(880.0,660.0),(25.0,230.0),(525.0,1000.0),
    (580.0,1175.0),(650.0,1130.0),(1605.0,620.0),(1220.0,580.0),
    (1465.0,200.0),(1530.0,5.0),(845.0,680.0),(725.0,370.0),
    (145.0,665.0),(415.0,635.0),(510.0,875.0),(560.0,365.0),
    (300.0,465.0),(520.0,585.0),(480.0,415.0),(835.0,625.0),
    (975.0,580.0),(1215.0,245.0),(1320.0,315.0),(1250.0,400.0),
    (660.0,180.0),(410.0,250.0),(420.0,555.0),(575.0,665.0),
    (1150.0,1160.0),(700.0,580.0),(685.0,595.0),(685.0,610.0),
    (770.0,610.0),(795.0,645.0),(720.0,635.0),(760.0,650.0),
    (475.0,960.0),(95.0,260.0),(875.0,920.0),(700.0,500.0),
    (555.0,815.0),(830.0,485.0),(1170.0,65.0),(830.0,610.0),
    (605.0,625.0),(595.0,360.0),(1340.0,725.0),(1740.0,245.0)
]
const OPTIMAL_BERLIN52 = 7542.0

function build_distance_matrix(pts; rounding=true)
    n = length(pts)
    D = zeros(Float64, n, n)
    for i in 1:n, j in 1:n
        i == j && continue
        d = hypot(pts[i][1] - pts[j][1], pts[i][2] - pts[j][2])
        D[i, j] = rounding ? round(d) : d
    end
    D
end

function tour_length(tour::Vector{Int}, D::Matrix{Float64})
    s = 0.0
    n = length(tour)
    @inbounds for k in 1:n-1
        s += D[tour[k], tour[k+1]]
    end
    s + D[tour[end], tour[1]]
end

function nearest_neighbor_tour(D::Matrix{Float64}; start::Int=1)
    n = size(D, 1)
    visited = falses(n)
    tour = Int[start]
    visited[start] = true
    cur = start
    for _ in 2:n
        nxt = 0
        best = Inf
        for j in 1:n
            visited[j] && continue
            if D[cur, j] < best
                best = D[cur, j]
                nxt = j
            end
        end
        push!(tour, nxt)
        visited[nxt] = true
        cur = nxt
    end
    return tour, tour_length(tour, D)
end

function best_nearest_neighbor_tour(D::Matrix{Float64})
    n = size(D, 1)
    best_tour = Int[]
    best_len = Inf
    for s in 1:n
        t, L = nearest_neighbor_tour(D; start=s)
        if L < best_len
            best_tour = t
            best_len = L
        end
    end
    return best_tour, best_len
end

function delta_2opt(tour::Vector{Int}, D::Matrix{Float64}, i::Int, j::Int)
    n = length(tour)
    (i == 1 && j == n) && return 0.0
    a = tour[i]
    b = tour[mod1(i + 1, n)]
    c = tour[j]
    d = tour[mod1(j + 1, n)]
    (D[a, c] + D[b, d]) - (D[a, b] + D[c, d])
end

function apply_2opt!(tour::Vector{Int}, i::Int, j::Int)
    tour[i+1:j] = reverse(tour[i+1:j])
    return tour
end

function two_opt_best_improvement!(tour::Vector{Int}, D::Matrix{Float64}; max_passes=200)
    n = length(tour)
    L = tour_length(tour, D)

    for _ in 1:max_passes
        bestΔ = 0.0
        best_i, best_j = 0, 0

        for i in 1:n-1, j in i+2:n
            (i == 1 && j == n) && continue
            Δ = delta_2opt(tour, D, i, j)
            if Δ < bestΔ
                bestΔ = Δ
                best_i, best_j = i, j
            end
        end

        if bestΔ < -1e-12
            apply_2opt!(tour, best_i, best_j)
            L += bestΔ
        else
            break
        end
    end

    return tour, L
end

function two_opt_strong!(tour::Vector{Int}, D::Matrix{Float64})
    return two_opt_best_improvement!(tour, D; max_passes=300)
end

function insertion_move(tour::Vector{Int}, i::Int, j::Int)
    n = length(tour)
    i == j && return copy(tour)
    t = copy(tour)
    city = t[i]
    deleteat!(t, i)
    if j > i
        insert!(t, j - 1, city)
    else
        insert!(t, j, city)
    end
    length(t) == n || error("Insertion move broke permutation")
    return t
end

function double_bridge_move(tour::Vector{Int}, rng)
    n = length(tour)
    cuts = sort(rand(rng, 2:n-1, 4))
    a, b, c, d = cuts
    p1 = tour[1:a]
    p2 = tour[a+1:b]
    p3 = tour[b+1:c]
    p4 = tour[c+1:d]
    p5 = tour[d+1:end]
    return vcat(p1, p3, p2, p4, p5)
end

function route_xy(pts, tour)
    cyc = vcat(tour, tour[1])
    xs = [pts[i][1] for i in cyc]
    ys = [pts[i][2] for i in cyc]
    return xs, ys
end

function gif_to_base64(tmpfile)
    b64 = open(io -> base64encode(read(io)), tmpfile, "r")
    rm(tmpfile; force=true)
    return b64
end

function display_side_by_side_gifs(left_b64, right_b64)
    html = """
    <div style="display:flex;gap:16px;align-items:flex-start;flex-wrap:nowrap;margin:8px 0 18px 0;">
      <img src="data:image/gif;base64,$left_b64" style="max-width:49%;height:auto;display:block;border:1px solid #ddd"/>
      <img src="data:image/gif;base64,$right_b64" style="max-width:49%;height:auto;display:block;border:1px solid #ddd"/>
    </div>
    """
    display(HTML(html))
end

make_geometric(α) = (T0, t, T) -> T * α
make_log(c)       = (T0, t, T) -> T0 / (1 + c * log(t + 2.0))
make_veryfast(β)  = (T0, t, T) -> T0 * exp(-β * t)

COLORS  = [:royalblue, :firebrick, :darkorange]
LSTYLES = [:solid, :dash, :dot]

function sa_tsp_single(
    D::Matrix{Float64}, T0, cool_fn, optimal_len;
    max_iter=450_000,
    k_stop=120_000,
    Tmin_frac=1e-7,
    seed=42,
    intensify_every=3000,
    strong_every=12000,
    reheat_every=18000,
    reheat_factor=0.65,
    save_stride=1500,
    use_doublebridge=true,
    start_mode=:best_nn,
    initial_strong_local_search=true
)
    rng = MersenneTwister(seed)
    n = size(D, 1)

    if start_mode == :best_nn
        x, L = best_nearest_neighbor_tour(D)
    elseif start_mode == :random
        x = randperm(rng, n)
        L = tour_length(x, D)
    else
        error("Unknown start_mode: $start_mode")
    end

    if initial_strong_local_search
        x, L = two_opt_strong!(x, D)
    end

    best_x = copy(x)
    best_val = L
    T = T0
    Tmin_stop = Tmin_frac * T0

    curr_hist = Vector{Float64}(undef, max_iter)
    best_hist = Vector{Float64}(undef, max_iter)
    T_hist    = Vector{Float64}(undef, max_iter)

    saved_tours      = Vector{Vector{Int}}()
    saved_best_tours = Vector{Vector{Int}}()
    saved_iter       = Int[]
    saved_curr       = Float64[]
    saved_best       = Float64[]

    no_improve = 0

    for t in 1:max_iter
        move_type = rand(rng)

        x_new = x
        L_new = L

        if move_type < 0.50
            i = rand(rng, 1:n-1)
            j = rand(rng, i+1:n)
            if !(i == 1 && j == n)
                Δ = delta_2opt(x, D, i, j)
                x_new = copy(x)
                apply_2opt!(x_new, i, j)
                L_new = L + Δ
            end

        elseif move_type < 0.75
            i = rand(rng, 1:n)
            j = rand(rng, 1:n)
            while j == i
                j = rand(rng, 1:n)
            end
            x_new = insertion_move(x, i, j)
            L_new = tour_length(x_new, D)

        elseif move_type < 0.90
            i = rand(rng, 1:n)
            j = rand(rng, 1:n)
            while j == i
                j = rand(rng, 1:n)
            end
            x_new = copy(x)
            x_new[i], x_new[j] = x_new[j], x_new[i]
            L_new = tour_length(x_new, D)

        else
            if use_doublebridge
                x_new = double_bridge_move(x, rng)
                L_new = tour_length(x_new, D)
            end
        end

        Δ = L_new - L

        if Δ <= 0 || rand(rng) < exp(-Δ / max(T, 1e-300))
            x = x_new
            L = L_new
        end

        if t % intensify_every == 0
            x2 = copy(x)
            x2, L2 = two_opt_best_improvement!(x2, D; max_passes=40)
            if L2 < L
                x = x2
                L = L2
            end
        end

        if t % strong_every == 0
            x2 = copy(x)
            x2, L2 = two_opt_strong!(x2, D)
            if L2 < L
                x = x2
                L = L2
            end
        end

        if L < best_val - 1e-12
            best_x = copy(x)
            best_val = L
            no_improve = 0
        else
            no_improve += 1
        end

        curr_hist[t] = L
        best_hist[t] = best_val
        T_hist[t]    = T

        if t == 1 || t % save_stride == 0 || t == max_iter
            push!(saved_tours, copy(x))
            push!(saved_best_tours, copy(best_x))
            push!(saved_iter, t)
            push!(saved_curr, L)
            push!(saved_best, best_val)
        end

        if t % reheat_every == 0 && no_improve > reheat_every ÷ 2 && best_val > optimal_len
            T = max(T, reheat_factor * T0)

            if use_doublebridge
                x = double_bridge_move(best_x, rng)
                x, L = two_opt_best_improvement!(x, D; max_passes=60)
            end

            if best_val <= optimal_len + max(150.0, 0.02 * optimal_len)
                x = double_bridge_move(best_x, rng)
                x, L = two_opt_strong!(x, D)
            end

            no_improve = 0
        else
            T = max(cool_fn(T0, t, T), 1e-12)
        end

        if best_val <= optimal_len + 1e-12
            if saved_iter[end] != t
                push!(saved_tours, copy(x))
                push!(saved_best_tours, copy(best_x))
                push!(saved_iter, t)
                push!(saved_curr, L)
                push!(saved_best, best_val)
            end
            return (
                best_tour        = best_x,
                best_len         = best_val,
                curr_hist        = curr_hist[1:t],
                best_hist        = best_hist[1:t],
                T_hist           = T_hist[1:t],
                saved_tours      = saved_tours,
                saved_best_tours = saved_best_tours,
                saved_iter       = saved_iter,
                saved_curr       = saved_curr,
                saved_best       = saved_best,
                iters_done       = t
            )
        end

        if no_improve >= k_stop && T <= Tmin_stop
            if saved_iter[end] != t
                push!(saved_tours, copy(x))
                push!(saved_best_tours, copy(best_x))
                push!(saved_iter, t)
                push!(saved_curr, L)
                push!(saved_best, best_val)
            end
            return (
                best_tour        = best_x,
                best_len         = best_val,
                curr_hist        = curr_hist[1:t],
                best_hist        = best_hist[1:t],
                T_hist           = T_hist[1:t],
                saved_tours      = saved_tours,
                saved_best_tours = saved_best_tours,
                saved_iter       = saved_iter,
                saved_curr       = saved_curr,
                saved_best       = saved_best,
                iters_done       = t
            )
        end
    end

    return (
        best_tour        = best_x,
        best_len         = best_val,
        curr_hist        = curr_hist,
        best_hist        = best_hist,
        T_hist           = T_hist,
        saved_tours      = saved_tours,
        saved_best_tours = saved_best_tours,
        saved_iter       = saved_iter,
        saved_curr       = saved_curr,
        saved_best       = saved_best,
        iters_done       = max_iter
    )
end

function construct_tour(start::Int, tau, eta, alpha, beta, n, rng)
    tour = [start]
    visited = falses(n)
    visited[start] = true
    current = start
    for _ in 1:n-1
        weights = zeros(Float64, n)
        total = 0.0
        for j in 1:n
            visited[j] && continue
            w = (tau[current, j]^alpha) * (eta[current, j]^beta)
            weights[j] = w
            total += w
        end
        nxt = 0
        if total <= 0
            candidates = [j for j in 1:n if !visited[j]]
            nxt = rand(rng, candidates)
        else
            r = rand(rng) * total
            acc = 0.0
            for j in 1:n
                visited[j] && continue
                acc += weights[j]
                if r <= acc
                    nxt = j
                    break
                end
            end
            nxt == 0 && (nxt = findfirst(!, visited))
        end
        push!(tour, nxt)
        visited[nxt] = true
        current = nxt
    end
    return tour
end

function run_aco(D::Matrix{Float64}, optimal_len;
                 num_ants=size(D, 1),
                 num_iters=300,
                 alpha=1.0,
                 beta=5.0,
                 rho=0.5,
                 Q=100.0,
                 tau0=1e-6,
                 elite_weight=3.0,
                 seed=42)

    rng = MersenneTwister(seed)
    n = size(D, 1)

    eta = zeros(Float64, n, n)
    for i in 1:n, j in 1:n
        i != j && (eta[i, j] = 1.0 / D[i, j])
    end

    tau = fill(tau0, n, n)
    for i in 1:n
        tau[i, i] = 0.0
    end

    best_tour = Int[]
    best_len  = Inf

    for _ in 1:num_iters
        iter_tours = Vector{Vector{Int}}()
        iter_lens  = Float64[]

        for _ in 1:num_ants
            tour = construct_tour(rand(rng, 1:n), tau, eta, alpha, beta, n, rng)
            tour, L = two_opt_best_improvement!(tour, D; max_passes=40)

            push!(iter_tours, tour)
            push!(iter_lens, L)

            if L < best_len
                best_len = L
                best_tour = copy(tour)
            end
        end

        tau .*= (1.0 - rho)

        for (tour, L) in zip(iter_tours, iter_lens)
            Δτ = Q / L
            for k in 1:n-1
                i, j = tour[k], tour[k+1]
                tau[i, j] += Δτ
                tau[j, i] += Δτ
            end
            i, j = tour[end], tour[1]
            tau[i, j] += Δτ
            tau[j, i] += Δτ
        end

        if !isempty(best_tour)
            Δe = elite_weight * Q / best_len
            for k in 1:n-1
                i, j = best_tour[k], best_tour[k+1]
                tau[i, j] += Δe
                tau[j, i] += Δe
            end
            i, j = best_tour[end], best_tour[1]
            tau[i, j] += Δe
            tau[j, i] += Δe
        end

        if best_len <= optimal_len + 1e-12
            return (
                best_tour   = best_tour,
                best_len    = best_len
            )
        end
    end

    return (
        best_tour   = best_tour,
        best_len    = best_len
    )
end

function route_xy(pts, tour)
    cyc = vcat(tour, tour[1])
    xs = [pts[i][1] for i in cyc]
    ys = [pts[i][2] for i in cyc]
    return xs, ys
end

function gif_to_base64(tmpfile)
    b64 = open(io -> base64encode(read(io)), tmpfile, "r")
    rm(tmpfile; force=true)
    return b64
end

function display_side_by_side_gifs(left_b64, right_b64)
    html = """
    <div style="display:flex;gap:16px;align-items:flex-start;flex-wrap:nowrap;margin:8px 0 18px 0;">
      <img src="data:image/gif;base64,$left_b64" style="max-width:49%;height:auto;display:block;border:1px solid #ddd"/>
      <img src="data:image/gif;base64,$right_b64" style="max-width:49%;height:auto;display:block;border:1px solid #ddd"/>
    </div>
    """
    display(HTML(html))
end

function make_tsp_route_gif(pts, saved_tours, saved_iter, saved_curr, saved_best, fname, sname, color;
                            fps=9, frame_idx=nothing, hold_last=8)
    city_x = [p[1] for p in pts]
    city_y = [p[2] for p in pts]
    xmin, xmax = minimum(city_x), maximum(city_x)
    ymin, ymax = minimum(city_y), maximum(city_y)
    padx = 0.06 * max(1.0, xmax - xmin)
    pady = 0.06 * max(1.0, ymax - ymin)

    N = length(saved_tours)
    frames = isnothing(frame_idx) ? collect(1:N) : collect(frame_idx)
    frames = clamp.(frames, 1, N)

    anim = @animate for k in vcat(frames, fill(frames[end], hold_last))
        tour = saved_tours[k]
        xs, ys = route_xy(pts, tour)

        p = scatter(city_x, city_y;
                    markersize=4, color=:black,
                    xlabel="x", ylabel="y",
                    title="$fname  |  $sname\nИтерация $(saved_iter[k])  |  current = $(round(saved_curr[k], digits=2))  |  best = $(round(saved_best[k], digits=2))",
                    legend=:topright, aspect_ratio=:equal,
                    xlims=(xmin-padx, xmax+padx), ylims=(ymin-pady, ymax+pady),
                    titlefontsize=9, guidefontsize=9, legendfontsize=7,
                    size=(620, 420), label="города")

        plot!(xs, ys; color=color, linewidth=2.2, alpha=0.95, label="текущий маршрут")
        scatter!([pts[tour[1]][1]], [pts[tour[1]][2]];
                 color=:green, markersize=7, markershape=:star5, label="старт")
    end

    tmp = tempname() * ".gif"
    gif(anim, tmp; fps=fps, show_msg=false)
    return gif_to_base64(tmp)
end

function make_best_route_gif(pts, saved_best_tours, saved_iter, saved_best, fname, sname, color;
                             fps=9, frame_idx=nothing, hold_last=8)
    city_x = [p[1] for p in pts]
    city_y = [p[2] for p in pts]
    xmin, xmax = minimum(city_x), maximum(city_x)
    ymin, ymax = minimum(city_y), maximum(city_y)
    padx = 0.06 * max(1.0, xmax - xmin)
    pady = 0.06 * max(1.0, ymax - ymin)

    N = length(saved_best_tours)
    frames = isnothing(frame_idx) ? collect(1:N) : collect(frame_idx)
    frames = clamp.(frames, 1, N)

    anim = @animate for k in vcat(frames, fill(frames[end], hold_last))
        tour = saved_best_tours[k]
        xs, ys = route_xy(pts, tour)

        p = scatter(city_x, city_y;
                    markersize=4, color=:black,
                    xlabel="x", ylabel="y",
                    title="$fname  |  $sname\nИтерация $(saved_iter[k])  |  best = $(round(saved_best[k], digits=2))",
                    legend=:topright, aspect_ratio=:equal,
                    xlims=(xmin-padx, xmax+padx), ylims=(ymin-pady, ymax+pady),
                    titlefontsize=9, guidefontsize=9, legendfontsize=7,
                    size=(620, 420), label="города")

        plot!(xs, ys; color=color, linewidth=2.8, alpha=0.98, label="лучший маршрут")
        scatter!([pts[tour[1]][1]], [pts[tour[1]][2]];
                 color=:green, markersize=7, markershape=:star5, label="старт")
    end

    tmp = tempname() * ".gif"
    gif(anim, tmp; fps=fps, show_msg=false)
    return gif_to_base64(tmp)
end

function draw_tour(pts, tour; title_str="", color=:royalblue, lw=2.0)
    xs, ys = route_xy(pts, tour)
    city_x = [p[1] for p in pts]
    city_y = [p[2] for p in pts]
    p = scatter(city_x, city_y;
                markersize=4, color=:black, label="города",
                xlabel="x", ylabel="y", aspect_ratio=:equal,
                title=title_str, legend=:topright)
    plot!(p, xs, ys; color=color, linewidth=lw, label="маршрут")
    return p
end

pts = collect(POINTS_BERLIN52)
D   = build_distance_matrix(pts; rounding=true)

println()
println("  Оптимум Berlin52 = 7542")
println()

SCHEDULES = [
    ("Геометрическое (α=0.99996)",  9000.0,  make_geometric(0.99996)),
    ("Логарифмическое (c=0.10)",   18000.0, make_log(0.10)),
    ("Сверхбыстрый (β=3.5e-5)",     8000.0,  make_veryfast(3.5e-5)),
]

SA_PARAMS = (450_000, 120_000, 1e-7, 3000, 12000, 18000, 0.65, 1500)

max_iter, k_stop, Tmin_frac, intensify_every, strong_every, reheat_every, reheat_factor, save_stride = SA_PARAMS

println("▶  Berlin52 — имитация отжига")
println("   max_iter=$max_iter   k_stop=$k_stop   Tmin_frac=$Tmin_frac")
println("   intensify_every=$intensify_every   strong_every=$strong_every   reheat_every=$reheat_every")
println("   " * "─"^72)

sa_results = NamedTuple[]

pc = plot(title="Berlin52 — текущее L на каждой итерации",
          xlabel="Итерация", ylabel="Текущее L",
          legend=:topright, titlefontsize=10, guidefontsize=8, legendfontsize=7)
hline!(pc, [OPTIMAL_BERLIN52]; label="L* = $OPTIMAL_BERLIN52",
       color=:green, linestyle=:dashdot, linewidth=1.5)

pb = plot(title="Berlin52 — лучшее L на каждой итерации",
          xlabel="Итерация", ylabel="Лучшее L",
          legend=:topright, titlefontsize=10, guidefontsize=8, legendfontsize=7)
hline!(pb, [OPTIMAL_BERLIN52]; label="L* = $OPTIMAL_BERLIN52",
       color=:green, linestyle=:dashdot, linewidth=1.5)

for (k, (sname, T0, cool_fn)) in enumerate(SCHEDULES)
    t_start = time()

    res = sa_tsp_single(
        D, T0, cool_fn, OPTIMAL_BERLIN52;
        max_iter=max_iter,
        k_stop=k_stop,
        Tmin_frac=Tmin_frac,
        seed=42 + 100k,
        intensify_every=intensify_every,
        strong_every=strong_every,
        reheat_every=reheat_every,
        reheat_factor=reheat_factor,
        save_stride=save_stride,
        use_doublebridge=true,
        start_mode=:best_nn,
        initial_strong_local_search=true
    )

    elapsed = time() - t_start
    gap = 100.0 * (res.best_len - OPTIMAL_BERLIN52) / OPTIMAL_BERLIN52

    @printf("   %-38s │ L*=%.2f  |Δ|=%.2f  │ %8d iter  │ %.4f с\n",
            sname, res.best_len, abs(res.best_len - OPTIMAL_BERLIN52),
            res.iters_done, elapsed)
    @printf("   %38s   gap = %+8.4f%%\n\n", "", gap)

    plot!(pc, 1:20:length(res.curr_hist), res.curr_hist[1:20:end];
          label=sname * @sprintf(" → %.1f", res.best_len),
          color=COLORS[k], linestyle=LSTYLES[k], linewidth=1.2, alpha=0.75)

    plot!(pb, 1:length(res.best_hist), res.best_hist;
          label=sname * @sprintf(" → %.1f", res.best_len),
          color=COLORS[k], linestyle=LSTYLES[k], linewidth=1.8, alpha=0.90)

    push!(sa_results, (
        sname   = sname,
        color   = COLORS[k],
        result  = res,
        elapsed = elapsed,
        gap     = gap
    ))
end

display(pc)
display(pb)

println()
for item in sa_results
    println("   ── $(item.sname)")

    Nframes_src = length(item.result.saved_iter)
    n_frames = min(45, Nframes_src)
    shared_idx = unique(round.(Int, range(1, Nframes_src; length=n_frames)))
    shared_idx[end] != Nframes_src && push!(shared_idx, Nframes_src)

    curr_b64 = make_tsp_route_gif(
        pts,
        item.result.saved_tours,
        item.result.saved_iter,
        item.result.saved_curr,
        item.result.saved_best,
        "Berlin52", item.sname, item.color;
        fps=9,
        frame_idx=shared_idx,
        hold_last=10
    )

    best_b64 = make_best_route_gif(
        pts,
        item.result.saved_best_tours,
        item.result.saved_iter,
        item.result.saved_best,
        "Berlin52", item.sname, item.color;
        fps=9,
        frame_idx=shared_idx,
        hold_last=10
    )

    display_side_by_side_gifs(curr_b64, best_b64)
end

println()
println("▶  Berlin52 — муравьиный алгоритм")
println("   " * "─"^72)

t_aco = time()
aco = run_aco(D, OPTIMAL_BERLIN52;
              num_ants=100,
              num_iters=300,
              alpha=1.0,
              beta=5.0,
              rho=0.5,
              Q=100.0,
              tau0=1e-6,
              elite_weight=3.0,
              seed=42)
elapsed_aco = time() - t_aco
gap_aco = 100.0 * (aco.best_len - OPTIMAL_BERLIN52) / OPTIMAL_BERLIN52

@printf("   %-38s │ L*=%.2f  |Δ|=%.2f  │ %.4f с\n",
        "ACO + 2-opt", aco.best_len, abs(aco.best_len - OPTIMAL_BERLIN52), elapsed_aco)
@printf("   %38s   gap = %+8.4f%%\n",
        "", gap_aco)

display(draw_tour(pts, aco.best_tour;
                  title_str=@sprintf("Berlin52 — ACO+2opt   L=%.2f", aco.best_len),
                  color=:purple, lw=2.0))

println()
@printf("  %-34s │ %-10s │ %-10s │ %-12s │ %-10s\n",
        "Метод", "Лучшее L", "gap,%", "iter", "time, с")
println("  " * "─"^34 * "┼" * "─"^12 * "┼" * "─"^12 * "┼" * "─"^14 * "┼" * "─"^12)

for item in sa_results
    @printf("  %-34s │ %-10.2f │ %+9.4f │ %-12d │ %-10.4f\n",
            item.sname, item.result.best_len, item.gap,
            item.result.iters_done, item.elapsed)
end

@printf("  %-34s │ %-10.2f │ %+9.4f │ %-12s │ %-10.4f\n",
        "ACO + 2-opt", aco.best_len, gap_aco, "—", elapsed_aco)
println("=" ^ 86)

println()

best_len_sa = minimum(x.result.best_len for x in sa_results)
cands = [x for x in sa_results if abs(x.result.best_len - best_len_sa) < 1e-12]
best_sa = cands[argmin([x.elapsed for x in cands])]

println("Лучший вариант SA : $(best_sa.sname)")
@printf("L = %.2f, gap = %+0.4f%%, итераций = %d, время = %.4f с\n",
        best_sa.result.best_len, best_sa.gap, best_sa.result.iters_done, best_sa.elapsed)
@printf("ACO + 2-opt: L = %.2f, gap = %+0.4f%%, время = %.4f с\n",
        aco.best_len, gap_aco, elapsed_aco)
```
