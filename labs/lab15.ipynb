```julia
using Plots
using Random
using Printf
using Base64
using Statistics

gr()

rastrigin(x::Vector{Float64}) =
    20.0 + x[1]^2 - 10cos(2π*x[1]) + x[2]^2 - 10cos(2π*x[2])

rosenbrock(x::Vector{Float64}) =
    100(x[2] - x[1]^2)^2 + (1 - x[1])^2

schwefel(x::Vector{Float64}) =
    418.9829*2 - x[1]*sin(sqrt(abs(x[1]))) - x[2]*sin(sqrt(abs(x[2])))

CONFIGS = [
    (rastrigin,  "Растригина",  (-5.12,   5.12),   [0.0,      0.0      ]),
    (rosenbrock, "Розенброка",  (-2.048,  2.048),  [1.0,      1.0      ]),
    (schwefel,   "Швефеля",     (-500.0,  500.0),  [420.9687, 420.9687 ]),
]

mutable struct Particle
    pos    :: Vector{Float64}
    vel    :: Vector{Float64}
    p_best :: Vector{Float64}
    pb_val :: Float64
end

function init_swarm(f, dim, num_particles, lb, ub)
    particles = [
        let pos = lb .+ rand(dim) .* (ub - lb),
            vel = (rand(dim) .* 2 .- 1) .* (ub - lb) .* 0.1,
            val = f(pos)
            Particle(pos, vel, copy(pos), val)
        end
        for _ in 1:num_particles
    ]
    gi         = argmin([p.pb_val for p in particles])
    g_best     = copy(particles[gi].p_best)
    g_best_val = particles[gi].pb_val
    return particles, g_best, g_best_val
end

function pso_classic(f, dim;
                     num_particles :: Int     = 30,
                     max_iter      :: Int     = 60,
                     w             :: Float64 = 0.729,
                     c1            :: Float64 = 1.496,
                     c2            :: Float64 = 1.496,
                     bounds                  = (-5.12, 5.12))

    lb, ub = Float64(bounds[1]), Float64(bounds[2])
    particles, g_best, g_best_val = init_swarm(f, dim, num_particles, lb, ub)
    all_positions = [hcat([copy(p.pos) for p in particles]...)]
    all_gbest     = [copy(g_best)]
    conv_history  = [g_best_val]

    for _ in 1:max_iter
        for p in particles
            rp = rand(dim); rg = rand(dim)
            p.vel = w  .* p.vel .+
                    c1 .* rp .* (p.p_best .- p.pos) .+
                    c2 .* rg .* (g_best   .- p.pos)
            p.pos .+= p.vel
            p.pos  .= clamp.(p.pos, lb, ub)
            val = f(p.pos)
            if val < p.pb_val; p.p_best = copy(p.pos); p.pb_val = val; end
        end
        bi = argmin([p.pb_val for p in particles])
        if particles[bi].pb_val < g_best_val
            g_best = copy(particles[bi].p_best); g_best_val = particles[bi].pb_val
        end
        push!(all_positions, hcat([copy(p.pos) for p in particles]...))
        push!(all_gbest, copy(g_best))
        push!(conv_history, g_best_val)
    end
    return all_positions, all_gbest, conv_history
end

function pso_lbest(f, dim;
                   num_particles :: Int     = 30,
                   max_iter      :: Int     = 60,
                   w             :: Float64 = 0.729,
                   c1            :: Float64 = 1.496,
                   c2            :: Float64 = 1.496,
                   neighborhood  :: Int     = 3,
                   bounds                  = (-5.12, 5.12))

    lb, ub = Float64(bounds[1]), Float64(bounds[2])
    particles, g_best, g_best_val = init_swarm(f, dim, num_particles, lb, ub)
    N = num_particles
    all_positions = [hcat([copy(p.pos) for p in particles]...)]
    all_gbest     = [copy(g_best)]
    conv_history  = [g_best_val]

    for _ in 1:max_iter
        for i in 1:N
            idxs   = [(i - neighborhood + k - 1 + N) % N + 1 for k in 0:2*neighborhood]
            l_best = particles[idxs[argmin([particles[j].pb_val for j in idxs])]]
            rp = rand(dim); rl = rand(dim)
            p  = particles[i]
            p.vel = w  .* p.vel .+
                    c1 .* rp .* (p.p_best      .- p.pos) .+
                    c2 .* rl .* (l_best.p_best  .- p.pos)
            p.pos .+= p.vel
            p.pos  .= clamp.(p.pos, lb, ub)
            val = f(p.pos)
            if val < p.pb_val; p.p_best = copy(p.pos); p.pb_val = val; end
        end
        bi = argmin([p.pb_val for p in particles])
        if particles[bi].pb_val < g_best_val
            g_best = copy(particles[bi].p_best); g_best_val = particles[bi].pb_val
        end
        push!(all_positions, hcat([copy(p.pos) for p in particles]...))
        push!(all_gbest, copy(g_best))
        push!(conv_history, g_best_val)
    end
    return all_positions, all_gbest, conv_history
end

function pso_inertia(f, dim;
                     num_particles :: Int     = 30,
                     max_iter      :: Int     = 60,
                     w_max         :: Float64 = 0.9,
                     w_min         :: Float64 = 0.4,
                     c1            :: Float64 = 1.496,
                     c2            :: Float64 = 1.496,
                     bounds                  = (-5.12, 5.12))

    lb, ub = Float64(bounds[1]), Float64(bounds[2])
    particles, g_best, g_best_val = init_swarm(f, dim, num_particles, lb, ub)
    all_positions = [hcat([copy(p.pos) for p in particles]...)]
    all_gbest     = [copy(g_best)]
    conv_history  = [g_best_val]

    for t in 1:max_iter
        w = w_max - (w_max - w_min) / max_iter * t
        for p in particles
            rp = rand(dim); rg = rand(dim)
            p.vel = w  .* p.vel .+
                    c1 .* rp .* (p.p_best .- p.pos) .+
                    c2 .* rg .* (g_best   .- p.pos)
            p.pos .+= p.vel
            p.pos  .= clamp.(p.pos, lb, ub)
            val = f(p.pos)
            if val < p.pb_val; p.p_best = copy(p.pos); p.pb_val = val; end
        end
        bi = argmin([p.pb_val for p in particles])
        if particles[bi].pb_val < g_best_val
            g_best = copy(particles[bi].p_best); g_best_val = particles[bi].pb_val
        end
        push!(all_positions, hcat([copy(p.pos) for p in particles]...))
        push!(all_gbest, copy(g_best))
        push!(conv_history, g_best_val)
    end
    return all_positions, all_gbest, conv_history
end

function pso_constriction(f, dim;
                          num_particles :: Int     = 30,
                          max_iter      :: Int     = 60,
                          c1            :: Float64 = 2.05,
                          c2            :: Float64 = 2.05,
                          kappa         :: Float64 = 1.0,
                          bounds                  = (-5.12, 5.12))

    lb, ub = Float64(bounds[1]), Float64(bounds[2])
    particles, g_best, g_best_val = init_swarm(f, dim, num_particles, lb, ub)
    phi = c1 + c2
    chi = 2kappa / abs(2 - phi - sqrt(phi^2 - 4phi))
    all_positions = [hcat([copy(p.pos) for p in particles]...)]
    all_gbest     = [copy(g_best)]
    conv_history  = [g_best_val]

    for _ in 1:max_iter
        for p in particles
            rp = rand(dim); rg = rand(dim)
            p.vel = chi .* (p.vel .+
                    c1 .* rp .* (p.p_best .- p.pos) .+
                    c2 .* rg .* (g_best   .- p.pos))
            p.pos .+= p.vel
            p.pos  .= clamp.(p.pos, lb, ub)
            val = f(p.pos)
            if val < p.pb_val; p.p_best = copy(p.pos); p.pb_val = val; end
        end
        bi = argmin([p.pb_val for p in particles])
        if particles[bi].pb_val < g_best_val
            g_best = copy(particles[bi].p_best); g_best_val = particles[bi].pb_val
        end
        push!(all_positions, hcat([copy(p.pos) for p in particles]...))
        push!(all_gbest, copy(g_best))
        push!(conv_history, g_best_val)
    end
    return all_positions, all_gbest, conv_history
end

function pso_teleport(f, dim;
                      num_particles :: Int     = 30,
                      max_iter      :: Int     = 60,
                      w             :: Float64 = 0.729,
                      c1            :: Float64 = 1.496,
                      c2            :: Float64 = 1.496,
                      teleport_prob :: Float64 = 0.075,
                      bounds                  = (-5.12, 5.12))

    lb, ub = Float64(bounds[1]), Float64(bounds[2])
    particles, g_best, g_best_val = init_swarm(f, dim, num_particles, lb, ub)
    all_positions = [hcat([copy(p.pos) for p in particles]...)]
    all_gbest     = [copy(g_best)]
    conv_history  = [g_best_val]

    for _ in 1:max_iter
        for p in particles
            if rand() < teleport_prob
                p.pos = lb .+ rand(dim) .* (ub - lb)
                p.vel .= 0.0
            else
                rp = rand(dim); rg = rand(dim)
                p.vel = w  .* p.vel .+
                        c1 .* rp .* (p.p_best .- p.pos) .+
                        c2 .* rg .* (g_best   .- p.pos)
                p.pos .+= p.vel
                p.pos  .= clamp.(p.pos, lb, ub)
            end
            val = f(p.pos)
            if val < p.pb_val; p.p_best = copy(p.pos); p.pb_val = val; end
        end
        bi = argmin([p.pb_val for p in particles])
        if particles[bi].pb_val < g_best_val
            g_best = copy(particles[bi].p_best); g_best_val = particles[bi].pb_val
        end
        push!(all_positions, hcat([copy(p.pos) for p in particles]...))
        push!(all_gbest, copy(g_best))
        push!(conv_history, g_best_val)
    end
    return all_positions, all_gbest, conv_history
end

function make_contour_grid(f, bounds; n=250)
    lb, ub = bounds
    xs = range(lb, ub, length=n)
    ys = range(lb, ub, length=n)
    Z  = [f([x, y]) for y in ys, x in xs]
    return xs, ys, clamp.(Z, quantile(vec(Z), 0.0), quantile(vec(Z), 0.97))
end

function legend_corner(true_min, bounds)
    lb, ub = bounds
    tm_nx = (true_min[1] - lb) / (ub - lb)
    tm_ny = (true_min[2] - lb) / (ub - lb)
    if tm_nx > 0.5 && tm_ny > 0.5;  return :bottomleft
    elseif tm_nx > 0.5;             return :topleft
    elseif tm_ny > 0.5;             return :bottomright
    else;                           return :topright
    end
end

function print_iter_table(label::String,
                           gbests::Vector{Vector{Float64}},
                           conv::Vector{Float64})
    n_iter = length(conv) - 1
    line   = repeat("-", 65)
    println()
    println("  ", label)
    println("  ", line)
    @printf("  %-6s  %14s  %14s  %18s\n", "iter", "x1", "x2", "f(g_best)")
    println("  ", line)
    prev_val = conv[1]
    for t in 0:n_iter
        gb       = gbests[t + 1]
        val      = conv[t + 1]
        improved = t > 0 && val < prev_val
        prev_val = val
        marker   = improved ? "  <-- новый минимум" : ""
        @printf("  %-6d  %14.6f  %14.6f  %18.8f%s\n",
                t, gb[1], gb[2], val, marker)
    end
    println("  ", line)
    @printf("  Итог: x* = (%.6f, %.6f)   f(x*) = %.8f\n",
            gbests[end][1], gbests[end][2], conv[end])
    println("  ", line)
    println()
end

function make_gif_b64(f, fname, mod_name, bounds, true_min, positions, gbests, conv; fps=10)
    lb, ub    = bounds
    xs, ys, Z = make_contour_grid(f, bounds)
    n_iter    = length(positions) - 1
    conv_ymax = max(conv[1] * 1.05, 1e-6)
    lpos      = legend_corner(true_min, bounds)

    anim = @animate for t in 0:n_iter
        pos = positions[t + 1]; gb = gbests[t + 1]

        p_main = contourf(xs, ys, Z;
                          levels=25, color=:viridis, linewidth=0, alpha=0.88,
                          xlims=(lb,ub), ylims=(lb,ub), aspect_ratio=:equal,
                          xlabel="x1", ylabel="x2",
                          title=@sprintf("%s · %s   итерация %d/%d   f(g*) = %.5f",
                                         mod_name, fname, t, n_iter, conv[t+1]),
                          titlefontsize=9, guidefontsize=8, tickfontsize=7,
                          colorbar=false, legend=lpos, legendfontsize=7,
                          foreground_color_legend=:white,
                          background_color_legend=RGBA(0,0,0,0.55))

        scatter!(p_main, pos[1,:], pos[2,:];
                 label="частицы", color=:white,
                 markerstrokecolor=:black, markerstrokewidth=0.5, markersize=4)
        scatter!(p_main, [true_min[1]], [true_min[2]];
                 label="истинный min", color=:yellow, markershape=:diamond,
                 markersize=9, markerstrokecolor=:black, markerstrokewidth=1)
        scatter!(p_main, [gb[1]], [gb[2]];
                 label="g_best", color=:red, markershape=:star5,
                 markersize=10, markerstrokecolor=:white, markerstrokewidth=1)

        p_conv = plot(0:t, conv[1:t+1];
                      xlabel="итерация", ylabel="f(g_best)",
                      title=@sprintf("сходимость · %s · %s", mod_name, fname),
                      color=:orangered, linewidth=2, legend=false,
                      xlims=(0,n_iter), ylims=(0,conv_ymax),
                      titlefontsize=8, guidefontsize=8, tickfontsize=7)

        plot(p_main, p_conv; layout=@layout([a{0.72w} b]), size=(900,440))
    end

    tmp = tempname() * ".gif"
    gif(anim, tmp; fps=fps, show_msg=false)
    b64 = open(io -> base64encode(read(io)), tmp, "r")
    rm(tmp)
    return b64
end

function make_png_b64(p)
    tmp = tempname() * ".png"
    savefig(p, tmp)
    b64 = open(io -> base64encode(read(io)), tmp, "r")
    rm(tmp)
    return b64
end

function display_gif(b64::String)
    display(HTML("""<img src="data:image/gif;base64,$(b64)" style="max-width:100%;display:block;margin:4px 0"/>"""))
end

function display_png(b64::String)
    display(HTML("""<img src="data:image/png;base64,$(b64)" style="max-width:100%;display:block;margin:4px 0"/>"""))
end

Random.seed!(42)

MAX_ITER      = 60
NUM_PARTICLES = 35

let phi = 2.05 + 2.05
    global chi_val = 2.0 * 1.0 / abs(2 - phi - sqrt(phi^2 - 4phi))
end

MODIFICATIONS = [
    ("Классический PSO",            pso_classic,      "w=0.729 · c1=c2=1.496"),
    ("LBEST (локальная топология)",  pso_lbest,        "w=0.729 · c1=c2=1.496 · K=3 соседей"),
    ("Линейная инерция w(t)",        pso_inertia,      "w: 0.9→0.4 · c1=c2=1.496"),
    ("Коэффициент сужения",          pso_constriction, @sprintf("c1=c2=2.05 · κ=1.0 · χ=%.6f", chi_val)),
    ("Телепортация",                 pso_teleport,     "w=0.729 · c1=c2=1.496 · p_teleport=7.5%"),
]

for (mod_name, mod_fn, mod_desc) in MODIFICATIONS
    println()
    println(repeat("=", 65))
    println("  ", mod_name, "  |  ", mod_desc)
    println(repeat("=", 65))

    for (f, fname, bounds, true_min) in CONFIGS
        positions, gbests, conv = mod_fn(f, 2;
            num_particles = NUM_PARTICLES,
            max_iter      = MAX_ITER,
            bounds        = bounds)

        print_iter_table("$(mod_name) · Функция $(fname)", gbests, conv)

        b64 = make_gif_b64(f, fname, mod_name, bounds, true_min,
                           positions, gbests, conv; fps=10)
        display_gif(b64)
    end
end

println()
println(repeat("=", 65))
println("  Сравнительные графики")
println(repeat("=", 65))

mod_labels = ["Классический", "LBEST", "Линейная инерция", "Сужение", "Телепортация"]
mod_fns    = [pso_classic, pso_lbest, pso_inertia, pso_constriction, pso_teleport]
mod_colors = [:steelblue, :orangered, :mediumseagreen, :purple, :gold]
particle_counts = [10, 100, 1000, 10000]
colors_n        = [:red, :orange, :steelblue, :mediumseagreen]
c2_values       = [0.5, 1.0, 1.496, 2.0, 2.5]
colors_c2       = [:purple, :steelblue, :mediumseagreen, :orange, :red]
teleport_probs  = [0.0, 0.025, 0.051, 0.075, 0.09, 0.15]
colors_tp       = [:black, :blue, :cyan, :green, :orange, :red]
labels_tp       = ["0.0% — нет", "2.5%", "5.1%", "7.5% — оптимум", "9%", "15% — максимум"]

for (cfg_idx, cfg_fname) in [(3, "Швефеля"), (1, "Растригина")]
    f, _, bounds, _ = CONFIGS[cfg_idx]

    Random.seed!(42)
    p_mods = plot(; title="Сравнение модификаций · $(cfg_fname)",
                  xlabel="Итерация", ylabel="f(g_best)",
                  size=(820,380), legend=:topright,
                  titlefontsize=11, guidefontsize=9, tickfontsize=8)
    for (i, fn) in enumerate(mod_fns)
        _, _, conv = fn(f, 2; num_particles=NUM_PARTICLES, max_iter=MAX_ITER, bounds=bounds)
        plot!(p_mods, 0:length(conv)-1, conv;
              label=mod_labels[i], color=mod_colors[i], linewidth=2)
    end
    display_png(make_png_b64(p_mods))

    Random.seed!(42)
    p_n = plot(; title="Влияние числа частиц · $(cfg_fname)",
               xlabel="Итерация", ylabel="f(g_best)",
               size=(820,360), legend=:topright,
               titlefontsize=11, guidefontsize=9, tickfontsize=8)
    for (i, N) in enumerate(particle_counts)
        _, _, conv = pso_classic(f, 2; num_particles=N, max_iter=MAX_ITER, bounds=bounds)
        plot!(p_n, 0:length(conv)-1, conv;
              label="N = $N", color=colors_n[i], linewidth=2)
    end
    display_png(make_png_b64(p_n))

    Random.seed!(42)
    p_c2 = plot(; title="Влияние параметра φg · $(cfg_fname)",
                xlabel="Итерация", ylabel="f(g_best)",
                size=(820,360), legend=:topright,
                titlefontsize=11, guidefontsize=9, tickfontsize=8)
    for (i, c2) in enumerate(c2_values)
        _, _, conv = pso_classic(f, 2;
            num_particles=NUM_PARTICLES, max_iter=MAX_ITER, c2=c2, bounds=bounds)
        plot!(p_c2, 0:length(conv)-1, conv;
              label="φg = $c2", color=colors_c2[i], linewidth=2)
    end
    display_png(make_png_b64(p_c2))

    Random.seed!(42)
    p_tp = plot(; title="Влияние вероятности телепортации · $(cfg_fname)",
                xlabel="Итерация", ylabel="f(g_best)",
                size=(820,380), legend=:topright,
                titlefontsize=11, guidefontsize=9, tickfontsize=8)
    for (i, prob) in enumerate(teleport_probs)
        _, _, conv = pso_teleport(f, 2;
            num_particles=NUM_PARTICLES, max_iter=MAX_ITER,
            teleport_prob=prob, bounds=bounds)
        plot!(p_tp, 0:length(conv)-1, conv;
              label=labels_tp[i], color=colors_tp[i], linewidth=2)
    end
    display_png(make_png_b64(p_tp))
end
```
