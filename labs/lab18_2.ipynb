```julia
using LinearAlgebra
using Random
using Printf
using Base64
using Plots
using Images
using ImageTransformations
using FileIO
using ColorTypes

gr()

const PHOTO_PATH = "photo.jpg"

const GA_ROWS = 28
const GA_COLS = 43

const OUT_H = 500
const OUT_W = 768

const POPSIZE = 140
const MAX_ITER = 2200
const ELITE_KEEP = 12
const TOURNAMENT_K = 5
const PMUT_CHILD = 0.55
const PGENE_START = 0.0009
const PGENE_END = 0.0002
const LOCAL_STEPS_PER_ITER = 6
const SEED = 14
const FPS = 10
const MAX_ANIM_FRAMES = 55

to_bits(x::Int) = reverse(digits(clamp(x, 0, 6), base = 2, pad = 3))
from_bits(b::Vector{Int}) = clamp(b[1] * 4 + b[2] * 2 + b[3], 0, 6)

euclid_dist(a::Vector{Int}, b::Vector{Int}) = norm(Float64.(a) .- Float64.(b))

function load_photo_color(path::String; rows::Int, cols::Int)
    img = load(path)
    img_small = imresize(img, (rows, cols))

    target = Vector{Int}(undef, rows * cols * 3)
    idx = 1

    for i in 1:rows, j in 1:cols
        px = RGB{Float64}(img_small[i, j])
        target[idx] = clamp(round(Int, red(px) * 6), 0, 6)
        target[idx + 1] = clamp(round(Int, green(px) * 6), 0, 6)
        target[idx + 2] = clamp(round(Int, blue(px) * 6), 0, 6)
        idx += 3
    end

    target
end

function vec_to_rgb(v::Vector{Int}, rows::Int, cols::Int)
    @assert length(v) == rows * cols * 3

    img = Matrix{RGB{Float64}}(undef, rows, cols)
    idx = 1

    for i in 1:rows, j in 1:cols
        r = v[idx] / 6.0
        g = v[idx + 1] / 6.0
        b = v[idx + 2] / 6.0
        img[i, j] = RGB(r, g, b)
        idx += 3
    end

    img
end

upscale(img::Matrix{RGB{Float64}}, out_h::Int, out_w::Int) = imresize(img, (out_h, out_w))

function gene_cross(a::Int, b::Int)
    ba = to_bits(a)
    bb = to_bits(b)
    p = rand(1:2)

    c1 = vcat(ba[1:p], bb[p + 1:end])
    c2 = vcat(bb[1:p], ba[p + 1:end])

    from_bits(c1), from_bits(c2)
end

function crossover_by_gene(a::Vector{Int}, b::Vector{Int})
    c1 = similar(a)
    c2 = similar(b)

    for i in eachindex(a)
        c1[i], c2[i] = gene_cross(a[i], b[i])
    end

    c1, c2
end

function mutate_gene(x::Int)
    b = to_bits(x)
    j = rand(1:3)
    b[j] = 1 - b[j]
    from_bits(b)
end

function mutate_density!(ch::Vector{Int}, p_gene::Float64)
    for i in eachindex(ch)
        if rand() < p_gene
            ch[i] = mutate_gene(ch[i])
        end
    end
end

function mutate_small!(ch::Vector{Int})
    k = rand(1:length(ch))
    old = ch[k]

    if rand() < 0.75
        delta = rand(Bool) ? 1 : -1
        ch[k] = clamp(old + delta, 0, 6)
    else
        ch[k] = mutate_gene(old)
    end
end

function tournament_select(population::Vector{Vector{Int}}, fits::Vector{Float64}, k::Int)
    inds = rand(1:length(population), k)
    best_idx = inds[1]
    best_fit = fits[best_idx]

    for idx in inds[2:end]
        if fits[idx] < best_fit
            best_fit = fits[idx]
            best_idx = idx
        end
    end

    population[best_idx]
end

function random_smooth_individual(rows::Int, cols::Int)
    img = Matrix{NTuple{3, Int}}(undef, rows, cols)

    block_r = max(2, rows ÷ 6)
    block_c = max(2, cols ÷ 6)

    for i in 1:rows, j in 1:cols
        bi = cld(i, block_r)
        bj = cld(j, block_c)
        Random.seed!(rand(1:10^9) + 97 * bi + 997 * bj)
        img[i, j] = (rand(0:6), rand(0:6), rand(0:6))
    end

    v = Vector{Int}(undef, rows * cols * 3)
    idx = 1

    for i in 1:rows, j in 1:cols
        r, g, b = img[i, j]
        v[idx] = r
        v[idx + 1] = g
        v[idx + 2] = b
        idx += 3
    end

    v
end

function init_population(popsize::Int, ngenes::Int, rows::Int, cols::Int)
    pop = Vector{Vector{Int}}(undef, popsize)
    n_random = popsize ÷ 2

    for i in 1:n_random
        pop[i] = rand(0:6, ngenes)
    end

    for i in (n_random + 1):popsize
        pop[i] = random_smooth_individual(rows, cols)
    end

    pop
end

function show_rgb(img::Matrix{RGB{Float64}}, ttl::String; w::Int = 350, h::Int = 240)
    plot(
        img,
        axis = false,
        ticks = false,
        title = ttl,
        titlefontsize = 9,
        size = (w, h),
        framestyle = :none
    )
end

function show_gif_inline_from_temp(anim; fps::Int = FPS)
    tmp = tempname() * ".gif"
    gif(anim, tmp; fps = fps, show_msg = false)
    b64 = open(io -> base64encode(read(io)), tmp, "r")
    rm(tmp; force = true)

    display(HTML("""
        <img src="data:image/gif;base64,$b64"
             style="max-width:100%;display:block;margin:8px 0"/>
    """))
end

function show_animation(res, target_img::Matrix{RGB{Float64}}; rows::Int, cols::Int, fps::Int = FPS)
    hist = res.history
    n_frames = min(MAX_ANIM_FRAMES, length(hist))
    step = max(1, length(hist) ÷ n_frames)

    frame_ids = collect(1:step:length(hist))
    append!(frame_ids, fill(length(hist), 8))

    anim = @animate for idx in frame_ids
        iter, best_vec, fit = hist[idx]
        cur_img = vec_to_rgb(best_vec, rows, cols)

        p1 = show_rgb(cur_img, "Лучшее решение | iter=$iter | err=$(round(fit, digits = 3))")
        p2 = show_rgb(target_img, "Эталон")

        plot(
            p1,
            p2,
            layout = (1, 2),
            size = (900, 320),
            plot_title = "Покоординатный ГА | евклидово расстояние",
            plot_titlefontsize = 10
        )
    end

    show_gif_inline_from_temp(anim; fps = fps)
end

function show_final(res, target_small::Matrix{RGB{Float64}}, rows::Int, cols::Int)
    best = upscale(vec_to_rgb(res.best, rows, cols), OUT_H, OUT_W)
    orig = upscale(target_small, OUT_H, OUT_W)

    p1 = show_rgb(orig, "Эталон $(OUT_H)×$(OUT_W)"; w = 500, h = 320)
    p2 = show_rgb(best, "Результат ГА\nerr=$(round(res.best_fit, digits = 3))"; w = 500, h = 320)

    display(
        plot(
            p1,
            p2,
            layout = (1, 2),
            size = (1080, 360),
            plot_title = "Восстановление цветного изображения"
        )
    )
end

function improve_best_local(best::Vector{Int}, target::Vector{Int}, nsteps::Int)
    cur = copy(best)
    cur_fit = euclid_dist(cur, target)

    for _ in 1:nsteps
        cand = copy(cur)
        mutate_small!(cand)
        f = euclid_dist(cand, target)

        if f < cur_fit
            cur = cand
            cur_fit = f
        end
    end

    cur, cur_fit
end

function run_ga_by_gene(
    target::Vector{Int};
    rows::Int,
    cols::Int,
    popsize::Int = POPSIZE,
    max_iter::Int = MAX_ITER,
    elite_keep::Int = ELITE_KEEP,
    tournament_k::Int = TOURNAMENT_K,
    pmut_child::Float64 = PMUT_CHILD,
    pgene_start::Float64 = PGENE_START,
    pgene_end::Float64 = PGENE_END,
    local_steps::Int = LOCAL_STEPS_PER_ITER,
    seed::Int = SEED
)
    Random.seed!(seed)

    ngenes = length(target)
    population = init_population(popsize, ngenes, rows, cols)

    history = Tuple{Int, Vector{Int}, Float64}[]

    best_ind = copy(population[1])
    best_fit = Inf
    stop_iter = max_iter

    for iter in 1:max_iter
        parent_fits = [euclid_dist(ind, target) for ind in population]
        order_par = sortperm(parent_fits)

        elites = [copy(population[i]) for i in order_par[1:elite_keep]]
        children = Vector{Vector{Int}}()

        t = (iter - 1) / max(1, max_iter - 1)
        p_gene = pgene_start * (1 - t) + pgene_end * t

        while length(children) < (popsize - elite_keep)
            p1 = tournament_select(population, parent_fits, tournament_k)
            p2 = tournament_select(population, parent_fits, tournament_k)

            c1, c2 = crossover_by_gene(p1, p2)

            if rand() < pmut_child
                mutate_density!(c1, p_gene)
            end

            if rand() < pmut_child
                mutate_density!(c2, p_gene)
            end

            push!(children, c1)

            if length(children) < (popsize - elite_keep)
                push!(children, c2)
            end
        end

        elite_best = copy(elites[1])
        elite_best, _ = improve_best_local(elite_best, target, local_steps)

        all_inds = vcat(population, children, elites, [elite_best])
        fits = [euclid_dist(ind, target) for ind in all_inds]
        order = sortperm(fits)

        population = [copy(all_inds[i]) for i in order[1:popsize]]

        cur_best = copy(population[1])
        cur_fit = euclid_dist(cur_best, target)

        if cur_fit < best_fit
            best_fit = cur_fit
            best_ind = copy(cur_best)
        end

        push!(history, (iter, cur_best, cur_fit))
        stop_iter = iter
    end

    @printf(
        "\nпокоординатный | евклидово | ошибка = %.4f | итерация = %d\n",
        best_fit,
        stop_iter
    )

    (
        best = best_ind,
        best_fit = best_fit,
        history = history,
        stop_iter = stop_iter
    )
end

function main()
    println("Загружаем фото: $PHOTO_PATH")
    println("Рабочая сетка ГА: $(GA_ROWS)×$(GA_COLS)")
    println("Финальное отображение: $(OUT_W)×$(OUT_H)")

    target_vec = load_photo_color(PHOTO_PATH; rows = GA_ROWS, cols = GA_COLS)
    target_small = vec_to_rgb(target_vec, GA_ROWS, GA_COLS)

    display(
        show_rgb(
            upscale(target_small, OUT_H, OUT_W),
            "Эталон (квантованный, $(OUT_W)×$(OUT_H))";
            w = 700,
            h = 420
        )
    )

    println("\n=== Запуск: только покоординатный вариант ===")

    res = run_ga_by_gene(
        target_vec;
        rows = GA_ROWS,
        cols = GA_COLS,
        popsize = POPSIZE,
        max_iter = MAX_ITER,
        elite_keep = ELITE_KEEP,
        tournament_k = TOURNAMENT_K,
        pmut_child = PMUT_CHILD,
        pgene_start = PGENE_START,
        pgene_end = PGENE_END,
        local_steps = LOCAL_STEPS_PER_ITER,
        seed = SEED
    )

    show_animation(res, target_small; rows = GA_ROWS, cols = GA_COLS, fps = FPS)
    show_final(res, target_small, GA_ROWS, GA_COLS)

    ph = plot(
        [f for (_, _, f) in res.history],
        xlabel = "Итерация",
        ylabel = "Евклидово расстояние",
        title = "Динамика сходимости",
        label = "by_gene / euclid",
        lw = 1.8
    )

    display(ph)

    res
end

main()
```
