```julia
using Random, Printf, LinearAlgebra, Statistics, Base64
using Plots
gr()

rastrigin(x::Vector{Float64})  = 20.0 + x[1]^2 - 10cos(2π*x[1]) + x[2]^2 - 10cos(2π*x[2])
rosenbrock(x::Vector{Float64}) = 100(x[2] - x[1]^2)^2 + (1 - x[1])^2
schwefel(x::Vector{Float64})   = 418.9829*2 - x[1]*sin(sqrt(abs(x[1]))) - x[2]*sin(sqrt(abs(x[2])))

function encode_var(x::Float64, lb::Float64, ub::Float64, nbits::Int)::Vector{Int}
    max_idx = 2^nbits - 1
    idx     = round(Int, clamp((x - lb) / (ub - lb), 0.0, 1.0) * max_idx)
    bits    = zeros(Int, nbits)
    tmp     = idx
    for i in nbits:-1:1
        bits[i] = tmp % 2
        tmp     = tmp ÷ 2
    end
    return bits
end

function decode_var(bits::Vector{Int}, lb::Float64, ub::Float64)::Float64
    idx = 0
    for b in bits
        idx = idx * 2 + b
    end
    return lb + (ub - lb) * idx / (2^length(bits) - 1)
end

function encode_individual(x::Vector{Float64}, lb::Float64, ub::Float64, nbits::Int)::Vector{Int}
    vcat([encode_var(xi, lb, ub, nbits) for xi in x]...)
end

function decode_individual(chrom::Vector{Int}, lb::Float64, ub::Float64, nbits::Int, ndim::Int)::Vector{Float64}
    [decode_var(chrom[(i-1)*nbits+1 : i*nbits], lb, ub) for i in 1:ndim]
end

function crossover(p1::Vector{Int}, p2::Vector{Int})
    point = rand(1:length(p1)-1)
    vcat(p1[1:point], p2[point+1:end]),
    vcat(p2[1:point], p1[point+1:end])
end

function mutate(chrom::Vector{Int}, p_mut::Float64)::Vector{Int}
    rand() >= p_mut && return copy(chrom)
    new_c = copy(chrom)
    new_c[rand(1:length(chrom))] ⊻= 1
    return new_c
end

function ga_step(pop::Vector{Vector{Int}}, n_pop::Int, p_mut::Float64, fitness_fn::Function)
    n           = length(pop)
    partner_ids = rand(1:n, n)
    offspring   = Vector{Int}[]
    for i in 1:n
        c1, c2 = crossover(pop[i], pop[partner_ids[i]])
        push!(offspring, c1, c2)
    end
    mutated  = [mutate(ch, p_mut) for ch in offspring]
    all_inds = vcat(pop, mutated)
    fits     = fitness_fn.(all_inds)
    order    = sortperm(fits)
    return all_inds[order[1:n_pop]], mutated, fits[order[1]]
end

function run_ga(f::Function, ndim::Int, lb::Float64, ub::Float64;
                n_pop::Int=100, n_gen::Int=300, p_mut::Float64=0.04,
                nbits::Int=20, n_restart::Int=3, seed::Int=42)

    decode  = chrom -> decode_individual(chrom, lb, ub, nbits, ndim)
    fitness = chrom -> f(decode(chrom))

    best_x_global = fill(Inf, ndim)
    best_f_global = Inf
    best_history  = nothing
    best_hist_f   = nothing

    for restart in 1:n_restart
        Random.seed!(seed + restart - 1)

        if restart == 1 || best_f_global == Inf
            pop = [encode_individual(lb .+ rand(ndim) .* (ub - lb), lb, ub, nbits) for _ in 1:n_pop]
        else
            radius    = (ub - lb) * 0.15
            half      = n_pop ÷ 2
            local_pop = [encode_individual(clamp.(best_x_global .+ (rand(ndim) .- 0.5) .* 2radius, lb, ub), lb, ub, nbits) for _ in 1:half]
            rand_pop  = [encode_individual(lb .+ rand(ndim) .* (ub - lb), lb, ub, nbits) for _ in 1:(n_pop - half)]
            pop       = vcat(local_pop, rand_pop)
        end

        history   = [(generation=0, initial=copy(pop), inter=Vector{Vector{Int}}(), final_pop=copy(pop))]
        history_f = [minimum(fitness.(pop))]

        for _ in 1:n_gen
            initial_this = copy(pop)
            pop, mutated, best_f = ga_step(pop, n_pop, p_mut, fitness)
            push!(history,   (generation=length(history), initial=initial_this, inter=mutated, final_pop=copy(pop)))
            push!(history_f, best_f)
        end

        cand_x = decode(pop[1])
        cand_f = f(cand_x)
        if cand_f < best_f_global
            best_f_global = cand_f
            best_x_global = cand_x
            best_history  = history
            best_hist_f   = history_f
        end
    end

    return best_x_global, best_f_global, best_hist_f, best_history, decode
end

function print_result(name, ndim, best_x, best_f, known_x, known_f)
    sep = "=" ^ 62
    println(sep)
    println("  Функция : $name")
    println(sep)
    println("  Найденное решение:")
    println("    x* = ", round.(best_x, digits=5))
    println("    f* = ", @sprintf("%.8f", best_f))
    println()
    println("  Истинный минимум:")
    println("    x  = $known_x")
    println("    f  = $known_f")
    println("  ||x* - x_true||₂ = ", @sprintf("%.6f", norm(best_x .- known_x)))
    println()
end

function make_and_display_gif(f2d::Function, lb::Float64, ub::Float64,
                               history::Vector, decode_fn::Function,
                               true_x1::Float64, true_x2::Float64, title_str::String;
                               fps::Int=5, npts::Int=180, legend_pos::Symbol=:topright)

    xs = range(lb, ub, length=npts)
    ys = range(lb, ub, length=npts)
    Z  = [f2d([x, y]) for y in ys, x in xs]

    z_sorted = sort(vec(Z))
    zlo      = z_sorted[1]
    zhi      = z_sorted[round(Int, 0.95 * length(z_sorted))]

    last_active = length(history)
    for g in 2:length(history)
        if history[g-1].final_pop == history[g].final_pop
            last_active = g
            break
        end
    end
    active_end = min(last_active + 5, length(history))

    n_frames  = min(40, active_end)
    step_auto = max(1, active_end ÷ n_frames)
    frame_ids = vcat(1:step_auto:active_end, fill(active_end, 10))

    anim = @animate for g in frame_ids
        state    = history[g]
        gen      = state.generation
        init_pts = [decode_fn(ch) for ch in state.initial]
        inter_pts = [decode_fn(ch) for ch in state.inter]
        fin_pts  = [decode_fn(ch) for ch in state.final_pop]

        init_x1  = [p[1] for p in init_pts];  init_x2  = [p[2] for p in init_pts]
        inter_x1 = [p[1] for p in inter_pts]; inter_x2 = [p[2] for p in inter_pts]
        fin_x1   = [p[1] for p in fin_pts];   fin_x2   = [p[2] for p in fin_pts]

        f_str = @sprintf("%.5f", minimum(f2d([fin_x1[i], fin_x2[i]]) for i in eachindex(fin_x1)))

        heatmap(xs, ys, clamp.(Z, zlo, zhi),
            color         = :viridis,
            title         = "$title_str — Поколение $gen  |  f* = $f_str",
            xlabel        = "x₁",
            ylabel        = "x₂",
            xlims         = (lb, ub),
            ylims         = (lb, ub),
            colorbar      = false,
            size          = (600, 520),
            titlefontsize = 10,
            legend        = legend_pos)

        scatter!([true_x1], [true_x2],
            marker=:diamond, markersize=12, color=:white,
            markerstrokecolor=:black, markerstrokewidth=1.5,
            label="x* истинный")

        scatter!(init_x1, init_x2,
            marker=:cross, markersize=8, color=:blue,
            markerstrokewidth=2, markerstrokecolor=:blue,
            label="Начальные особи")

        if !isempty(inter_pts)
            scatter!(inter_x1, inter_x2,
                marker=:circle, markersize=5, color=:red,
                markerstrokewidth=0,
                label="После скрещивания и мутации")
        end

        scatter!(fin_x1, fin_x2,
            marker=:circle, markersize=7, color=:black,
            markerstrokewidth=0,
            label="После отбора")
    end

    tmp = tempname() * ".gif"
    gif(anim, tmp; fps=fps, show_msg=false)
    b64 = open(io -> base64encode(read(io)), tmp, "r")
    rm(tmp)
    display(HTML("""<img src="data:image/gif;base64,$(b64)"
        style="max-width:100%;display:block;margin:8px 0"/>"""))
end

bx_r,  bf_r,  hist_f_r,  hist_r,  dec_r  = run_ga(rastrigin,  2, -5.12,  5.12;   n_pop=150, n_gen=600,  p_mut=0.02,  nbits=22, n_restart=5, seed=42)
print_result("Растригин", 2, bx_r, bf_r, [0.0, 0.0], 0.0)

bx_rb, bf_rb, hist_f_rb, hist_rb, dec_rb = run_ga(rosenbrock, 2, -2.048, 2.048;  n_pop=200, n_gen=1000, p_mut=0.015, nbits=22, n_restart=5, seed=7)
print_result("Розенброк", 2, bx_rb, bf_rb, [1.0, 1.0], 0.0)

bx_s,  bf_s,  hist_f_s,  hist_s,  dec_s  = run_ga(schwefel,   2, -500.0, 500.0;  n_pop=150, n_gen=700,  p_mut=0.03,  nbits=22, n_restart=5, seed=13)
print_result("Швефель", 2, bx_s, bf_s, [420.9687, 420.9687], 0.0)

p_conv = plot(
    title="Сходимость ГА", xlabel="Поколение", ylabel="f*",
    legend=:topright, size=(820, 420), yscale=:log10, minorgrid=true)
plot!(p_conv, 0:length(hist_f_r)-1,  max.(hist_f_r,  1e-12), label="Растригин", lw=2, color=:royalblue)
plot!(p_conv, 0:length(hist_f_rb)-1, max.(hist_f_rb, 1e-12), label="Розенброк", lw=2, color=:crimson)
plot!(p_conv, 0:length(hist_f_s)-1,  max.(hist_f_s,  1e-12), label="Швефель",   lw=2, color=:darkorange)
display(p_conv)

let _hist_r=hist_r, _dec_r=dec_r, _hist_rb=hist_rb, _dec_rb=dec_rb, _hist_s=hist_s, _dec_s=dec_s

    println("── Растригин " * "─"^48)
    make_and_display_gif(rastrigin,  -5.12,  5.12,  _hist_r,  _dec_r,  0.0,       0.0,       "Растригин"; fps=5)

    println("── Розенброк " * "─"^48)
    make_and_display_gif(rosenbrock, -2.048, 2.048, _hist_rb, _dec_rb, 1.0,       1.0,       "Розенброк"; fps=5)

    println("── Швефель " * "─"^50)
    make_and_display_gif(schwefel,   -500.0, 500.0, _hist_s,  _dec_s,  420.9687,  420.9687,  "Швефель";   fps=5, legend_pos=:bottomright)
end
```
