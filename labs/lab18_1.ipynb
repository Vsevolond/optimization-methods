```julia
using LinearAlgebra, Random, Printf, Base64, Plots
gr()

to_bits(x)   = reverse(digits(clamp(x, 0, 6), base=2, pad=3))
from_bits(b) = clamp(b[1]*4 + b[2]*2 + b[3], 0, 6)

function flatten_bits(v::Vector{Int})
    out = Int[]
    sizehint!(out, 3 * length(v))
    for x in v
        append!(out, to_bits(x))
    end
    return out
end

bits_to_vec(bits::Vector{Int}) = [from_bits(bits[3i-2:3i]) for i in 1:length(bits)÷3]

bool_dist(a, b)   = sum(flatten_bits(a) .!= flatten_bits(b))
euclid_dist(a, b) = norm(Float64.(a) .- Float64.(b))

function calc_dist(a, b, typ::Symbol)
    if typ == :bool
        return bool_dist(a, b)
    else
        return euclid_dist(a, b)
    end
end

function make_target(n::Int)
    img = [mod(i + j, 6) + 1 for i in 1:n, j in 1:n]
    for i in 1:n
        img[i, i]     = 0
        img[i, n-i+1] = 0
    end
    return img
end

img_to_vec(img)  = vec(copy(img))
vec_to_img(v, n) = reshape(copy(v), n, n)

function crossover_multipoint(a::Vector{Int}, b::Vector{Int}; n_points::Int=3)
    L      = length(a)
    points = sort(randperm(L-1)[1:min(n_points, L-1)])
    c1, c2 = copy(a), copy(b)
    swap = false
    prev = 1
    for pt in vcat(points, L)
        if swap
            c1[prev:pt] = b[prev:pt]
            c2[prev:pt] = a[prev:pt]
        end
        swap = !swap
        prev = pt + 1
    end
    return c1, c2
end

function crossover_shuffle(a::Vector{Int}, b::Vector{Int})
    L      = length(a)
    perm   = randperm(L)
    sa, sb = a[perm], b[perm]
    pt     = rand(1:L-1)
    sc1    = vcat(sa[1:pt], sb[pt+1:end])
    sc2    = vcat(sb[1:pt], sa[pt+1:end])
    ip     = invperm(perm)
    return sc1[ip], sc2[ip]
end

function mutate_bits(bits::Vector{Int}, pm::Float64)::Vector{Int}
    out = copy(bits)
    for i in eachindex(out)
        if rand() < pm
            out[i] ⊻= 1
        end
    end
    return out
end

function cross_mutate_whole(p1, p2, pm)
    ba1, ba2 = flatten_bits(p1), flatten_bits(p2)
    bc1, bc2 = rand() < 0.5 ? crossover_multipoint(ba1, ba2; n_points=3) :
                               crossover_shuffle(ba1, ba2)
    return bits_to_vec(mutate_bits(bc1, pm)),
           bits_to_vec(mutate_bits(bc2, pm))
end

function crossover_twopoint_gene(a::Vector{Int}, b::Vector{Int})
    L      = length(a)
    p1, p2 = sort(randperm(L)[1:2])
    c1     = vcat(a[1:p1-1], b[p1:p2], a[p2+1:end])
    c2     = vcat(b[1:p1-1], a[p1:p2], b[p2+1:end])
    return c1, c2
end

function crossover_uniform_gene(a::Vector{Int}, b::Vector{Int})
    mask = rand(Bool, length(a))
    c1   = [mask[i] ? a[i] : b[i] for i in eachindex(a)]
    c2   = [mask[i] ? b[i] : a[i] for i in eachindex(a)]
    return c1, c2
end

function mutate_genes_full(chrom::Vector{Int}, pm::Float64)::Vector{Int}
    out = copy(chrom)
    L   = length(out)

    for i in eachindex(out)
        if rand() < pm
            out[i] = rand(0:6)
        end
    end

    if rand() < pm * L * 0.5
        i, j = randperm(L)[1:2]
        out[i], out[j] = out[j], out[i]
    end

    if rand() < pm * L
        k    = rand(1:L)
        b    = to_bits(out[k])
        b[rand(1:3)] ⊻= 1
        out[k] = from_bits(b)
    end

    return out
end

function cross_mutate_gene(p1, p2, pm)
    c1, c2 = rand() < 0.5 ? crossover_twopoint_gene(p1, p2) :
                             crossover_uniform_gene(p1, p2)
    return mutate_genes_full(c1, pm), mutate_genes_full(c2, pm)
end

function tournament_select(population, fits; k::Int=3)
    idxs   = randperm(length(population))[1:k]
    best_i = idxs[argmin(fits[idxs])]
    return population[best_i]
end

function run_ga(n::Int; variant::Symbol, typ::Symbol,
                popsize::Int, max_iter::Int,
                pm_init::Float64, pm_min::Float64, pm_max::Float64,
                patience::Int=30, restart_patience::Int=80,
                eps::Float64=0.0, seed::Int=14, tour_k::Int=3)

    Random.seed!(seed)
    target_img = make_target(n)
    target     = img_to_vec(target_img)
    population = [rand(0:6, n*n) for _ in 1:popsize]

    best_ind   = copy(population[1])
    best_fit   = Inf
    history    = Tuple{Int, Vector{Int}, Float64}[]
    stop_iter  = max_iter
    pm         = pm_init
    no_improve = 0
    prev_best  = Inf

    for iter in 1:max_iter
        fits  = [calc_dist(x, target, typ) for x in population]
        order = sortperm(fits)

        elites  = [copy(population[order[1]]), copy(population[order[2]])]
        cur_fit = fits[order[1]]

        if cur_fit < best_fit
            best_fit = cur_fit
            best_ind = copy(population[order[1]])
        end

        push!(history, (iter, copy(population[order[1]]), cur_fit))

        if cur_fit <= eps
            stop_iter = iter
            break
        end

        if cur_fit < prev_best - 1e-9
            no_improve = 0
            prev_best  = cur_fit
            pm         = max(pm_min, pm * 0.95)
        else
            no_improve += 1
            if no_improve >= patience
                pm = min(pm_max, pm * 1.3)
            end
        end

        if no_improve >= restart_patience
            half       = popsize ÷ 2
            kept       = [copy(population[order[i]]) for i in 1:half]
            new_rand   = [rand(0:6, n*n) for _ in 1:popsize-half]
            population = vcat(kept, new_rand)
            no_improve = 0
            pm         = pm_init
            continue
        end

        children = Vector{Vector{Int}}()
        while length(children) < popsize
            p1 = tournament_select(population, fits; k=tour_k)
            p2 = tournament_select(population, fits; k=tour_k)
            c1, c2 = if variant == :whole
                cross_mutate_whole(p1, p2, pm)
            else
                cross_mutate_gene(p1, p2, pm)
            end
            push!(children, c1)
            if length(children) < popsize
                push!(children, c2)
            end
        end

        cf         = [calc_dist(x, target, typ) for x in children]
        cord       = sortperm(cf)
        population = vcat(elites, [copy(children[i]) for i in cord[1:popsize-2]])
    end

    vname = variant == :whole ? "сплошной вектор" : "покоординатный"
    tname = typ     == :bool  ? "булево расст."   : "евклидово расст."
    @printf("\n%dx%d | %s | %s | ошибка: %.4f | итерация: %d\n",
            n, n, vname, tname, best_fit, stop_iter)

    return (n=n, target_img=target_img, target_vec=target,
            best=best_ind, best_fit=best_fit,
            history=history, stop_iter=stop_iter,
            variant=variant, typ=typ)
end

function img_heatmap(mat::Matrix{Int}, ttl::String; fs=9)
    heatmap(mat[end:-1:1, :],
            c=cgrad(:grays), clims=(0, 6),
            colorbar=false, aspect_ratio=1,
            framestyle=:box, xticks=false, yticks=false,
            title=ttl, titlefontsize=fs)
end

function show_gif(res; fps::Int=5)
    n     = res.n
    hist  = res.history
    vname = res.variant == :whole ? "сплошной вектор" : "покоординатный"
    tname = res.typ     == :bool  ? "булево расст."   : "евклидово расст."

    n_frames  = min(40, length(hist))
    step      = max(1, length(hist) ÷ n_frames)
    frame_ids = vcat(1:step:length(hist), fill(length(hist), 8))

    anim = @animate for idx in frame_ids
        iter, bv, fit = hist[idx]
        p1 = img_heatmap(vec_to_img(bv, n),
                         "Лучшее  iter=$(iter)  err=$(round(fit, digits=2))")
        p2 = img_heatmap(res.target_img, "Эталон")
        plot(p1, p2, layout=(1, 2), size=(700, 340),
             plot_title="$(n)×$(n) | $(vname) | $(tname)",
             plot_titlefontsize=9)
    end

    tmp = tempname() * ".gif"
    gif(anim, tmp; fps=fps, show_msg=false)
    b64 = open(io -> base64encode(read(io)), tmp, "r")
    rm(tmp)
    display(HTML("""<img src="data:image/gif;base64,$(b64)"
        style="max-width:100%;display:block;margin:8px 0"/>"""))
end

function show_final(res)
    vname = res.variant == :whole ? "сплошной вектор" : "покоординатный"
    tname = res.typ     == :bool  ? "булево расст."   : "евклидово расст."
    p1 = img_heatmap(res.target_img, "Эталон $(res.n)×$(res.n)"; fs=10)
    p2 = img_heatmap(vec_to_img(res.best, res.n),
                     "Результат | $(vname) | $(tname)\n" *
                     "err=$(round(res.best_fit, digits=2))  iter=$(res.stop_iter)"; fs=9)
    display(plot(p1, p2, layout=(1, 2), size=(640, 320)))
end

configs = [
    (n=3,  popsize=40,  max_iter=300,
     pm_init=0.03,  pm_min=0.005, pm_max=0.15,
     patience=20,   restart_patience=60),

    (n=5,  popsize=80,  max_iter=600,
     pm_init=0.02,  pm_min=0.003, pm_max=0.12,
     patience=25,   restart_patience=80),

    (n=9,  popsize=150, max_iter=1200,
     pm_init=0.01,  pm_min=0.002, pm_max=0.08,
     patience=30,   restart_patience=100),

    (n=21, popsize=300, max_iter=2500,
     pm_init=0.005, pm_min=0.001, pm_max=0.04,
     patience=40,   restart_patience=150),
]

println("=== Вариант 1: сплошной вектор, булево расстояние ===")
results_v1 = []
for cfg in configs
    res = run_ga(cfg.n; variant=:whole, typ=:bool,
                 popsize=cfg.popsize, max_iter=cfg.max_iter,
                 pm_init=cfg.pm_init, pm_min=cfg.pm_min, pm_max=cfg.pm_max,
                 patience=cfg.patience, restart_patience=cfg.restart_patience,
                 eps=0.0)
    show_gif(res)
    show_final(res)
    push!(results_v1, res)
end

println("\n=== Вариант 2: покоординатный, евклидово расстояние ===")
results_v2 = []
for cfg in configs
    res = run_ga(cfg.n; variant=:by_gene, typ=:euclid,
                 popsize=cfg.popsize, max_iter=cfg.max_iter,
                 pm_init=cfg.pm_init, pm_min=cfg.pm_min, pm_max=cfg.pm_max,
                 patience=cfg.patience, restart_patience=cfg.restart_patience,
                 eps=1e-9)
    show_gif(res)
    show_final(res)
    push!(results_v2, res)
end

ph = plot(title="Динамика расстояния",
          xlabel="Итерация", ylabel="Расстояние", legend=:topright)
for res in vcat(results_v1, results_v2)
    vname = res.variant == :whole ? "сплош." : "покоорд."
    tname = res.typ     == :bool  ? "булево" : "евклид"
    fits  = [f for (_, _, f) in res.history]
    plot!(ph, fits, label="$(res.n)×$(res.n) $(vname)/$(tname)", lw=1.2)
end
display(ph)
```
